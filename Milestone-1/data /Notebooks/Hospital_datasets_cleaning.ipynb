```python
input_folder = "/content/drive/MyDrive/RAW"
output_folder = "/content/drive/MyDrive/RAW/Cleaned"

os.makedirs(output_folder, exist_ok=True)


# ------------------------------------------------------------
# 2. LOAD DATASETS
# ------------------------------------------------------------

patients = pd.read_csv(
    os.path.join(input_folder, "patients.csv"),
    low_memory=False
)

readmission = pd.read_csv(
    os.path.join(input_folder, "Healthcare Data Analysis for readmission.csv"),
    low_memory=False
)

inpatient = pd.read_csv(
    os.path.join(
        input_folder,
        "Hospital_Inpatient_Discharges__SPARCS_De-Identified___2021_20231012.csv"
    ),
    low_memory=False
)


# ============================================================
# DATASET 1: PATIENTS
# ============================================================

print("Cleaning patients.csv...")

# Standardize column names
patients.columns = patients.columns.str.strip().str.lower()

# Standardize text columns
patients["name"] = patients["name"].astype(str).str.strip()
patients["service"] = patients["service"].astype(str).str.strip().str.lower()

# Convert dates
patients["arrival_date"] = pd.to_datetime(
    patients["arrival_date"],
    errors="coerce"
)

patients["departure_date"] = pd.to_datetime(
    patients["departure_date"],
    errors="coerce"
)

# Remove exact duplicate rows
patients = patients.drop_duplicates()

# Check invalid values
patients.loc[
    (patients["age"] < 0) | (patients["age"] > 120),
    "age"
] = pd.NA

patients.loc[
    (patients["satisfaction"] < 0) | (patients["satisfaction"] > 100),
    "satisfaction"
] = pd.NA

print("Patients cleaned:", patients.shape)


# ============================================================
# DATASET 2: READMISSION / HMIS
# ============================================================

print("\nCleaning Healthcare Data Analysis for readmission.csv...")

# Standardize column names
readmission.columns = readmission.columns.str.strip().str.lower()

# Standardize text columns
text_columns = [
    "hospital_name",
    "hospital_ward",
    "patient_gender",
    "patient_race",
    "patient_first_initial",
    "patient_last_name",
    "department_referral",
    "time_slot",
    "doctor_name",
    "doctor_specialty",
    "patient_disease",
    "discharge_status"
]

for col in text_columns:
    readmission[col] = (
        readmission[col]
        .astype(str)
        .str.strip()
    )

# Standardize categorical values
readmission["hospital_ward"] = (
    readmission["hospital_ward"]
    .str.lower()
    .str.strip()
)

readmission["department_referral"] = (
    readmission["department_referral"]
    .str.lower()
    .str.strip()
)

readmission["patient_gender"] = (
    readmission["patient_gender"]
    .str.lower()
    .str.strip()
)

readmission["patient_race"] = (
    readmission["patient_race"]
    .str.lower()
    .str.strip()
)

readmission["discharge_status"] = (
    readmission["discharge_status"]
    .str.lower()
    .str.strip()
)

# Convert dates
date_columns = [
    "admission_date",
    "patient_checkin_date",
    "patient_checkout_date"
]

for col in date_columns:
    readmission[col] = pd.to_datetime(
        readmission[col],
        errors="coerce",
        dayfirst=True
    )

# Remove exact duplicates
readmission = readmission.drop_duplicates()

# Standardize IDs
readmission["hospital_id"] = (
    pd.to_numeric(readmission["hospital_id"], errors="coerce")
    .astype("Int64")
)

readmission["patient_id"] = (
    pd.to_numeric(readmission["patient_id"], errors="coerce")
    .astype("Int64")
)

readmission["doctor_id"] = (
    pd.to_numeric(readmission["doctor_id"], errors="coerce")
    .astype("Int64")
)

# Validate numeric ranges
readmission.loc[
    (readmission["patient_age"] < 0) |
    (readmission["patient_age"] > 120),
    "patient_age"
] = pd.NA

readmission.loc[
    readmission["patient_sat_score"] < 0,
    "patient_sat_score"
] = pd.NA

readmission.loc[
    readmission["patient_length_of_stay"] < 0,
    "patient_length_of_stay"
] = pd.NA

readmission.loc[
    readmission["patient_waittime"] < 0,
    "patient_waittime"
] = pd.NA

# Ensure readmission is 0/1
readmission["readmission"] = pd.to_numeric(
    readmission["readmission"],
    errors="coerce"
)

readmission.loc[
    ~readmission["readmission"].isin([0, 1]),
    "readmission"
] = pd.NA

print("Readmission dataset cleaned:", readmission.shape)


# ============================================================
# DATASET 3: INPATIENT DISCHARGES
# ============================================================

print("\nCleaning Hospital Inpatient Discharges dataset...")

# Standardize column names
inpatient.columns = (
    inpatient.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

# ------------------------------------------------------------
# Remove exact duplicate rows
# ------------------------------------------------------------

before = len(inpatient)

inpatient = inpatient.drop_duplicates()

after = len(inpatient)

print("Duplicate rows removed:", before - after)


# ------------------------------------------------------------
# Standardize text columns
# ------------------------------------------------------------

text_columns = inpatient.select_dtypes(
    include=["object"]
).columns

for col in text_columns:
    inpatient[col] = (
        inpatient[col]
        .astype("string")
        .str.strip()
    )


# ------------------------------------------------------------
# Convert numeric columns
# ------------------------------------------------------------

numeric_columns = [
    "operating_certificate_number",
    "permanent_facility_id",
    "apr_drg_code",
    "apr_mdc_code",
    "apr_severity_of_illness_code"
]

for col in numeric_columns:
    inpatient[col] = pd.to_numeric(
        inpatient[col],
        errors="coerce"
    )


# ------------------------------------------------------------
# Length of Stay
# ------------------------------------------------------------

inpatient["length_of_stay"] = pd.to_numeric(
    inpatient["length_of_stay"],
    errors="coerce"
)

inpatient.loc[
    inpatient["length_of_stay"] < 0,
    "length_of_stay"
] = pd.NA


# ------------------------------------------------------------
# Financial columns
# Remove commas and convert to numeric
# ------------------------------------------------------------

inpatient["total_charges"] = pd.to_numeric(
    inpatient["total_charges"]
    .astype(str)
    .str.replace(",", "", regex=False),
    errors="coerce"
)

inpatient["total_costs"] = pd.to_numeric(
    inpatient["total_costs"]
    .astype(str)
    .str.replace(",", "", regex=False),
    errors="coerce"
)


# ------------------------------------------------------------
# Birth Weight
# ------------------------------------------------------------

inpatient["birth_weight"] = pd.to_numeric(
    inpatient["birth_weight"],
    errors="coerce"
)


# ------------------------------------------------------------
# Standardize categorical values
# ------------------------------------------------------------

categorical_columns = [
    "gender",
    "race",
    "ethnicity",
    "type_of_admission",
    "patient_disposition",
    "apr_risk_of_mortality",
    "apr_medical_surgical_description",
    "payment_typology_1",
    "payment_typology_2",
    "payment_typology_3",
    "emergency_department_indicator"
]

for col in categorical_columns:
    inpatient[col] = (
        inpatient[col]
        .astype("string")
        .str.strip()
        .str.lower()
    )


# ------------------------------------------------------------
# Missing values
# ------------------------------------------------------------
# Do NOT replace all missing values with 0.
# Keep genuine missing information as NA.
#
# Only numerical financial values are retained as numeric;
# categorical missing values remain missing.
# ------------------------------------------------------------


# ============================================================
# SAVE CLEANED DATASETS
# ============================================================

patients.to_csv(
    os.path.join(output_folder, "patients_cleaned.csv"),
    index=False
)

readmission.to_csv(
    os.path.join(output_folder, "readmission_cleaned.csv"),
    index=False
)

inpatient.to_csv(
    os.path.join(output_folder, "inpatient_discharges_cleaned.csv"),
    index=False
)


# ============================================================
# CLEANING SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("DATA CLEANING COMPLETED")
print("=" * 70)

print("\nPatients:")
print("Rows:", len(patients))
print("Columns:", len(patients.columns))
print("Duplicates:", patients.duplicated().sum())

print("\nReadmission:")
print("Rows:", len(readmission))
print("Columns:", len(readmission.columns))
print("Duplicates:", readmission.duplicated().sum())

print("\nInpatient Discharges:")
print("Rows:", len(inpatient))
print("Columns:", len(inpatient.columns))
print("Duplicates:", inpatient.duplicated().sum())

print("\nCleaned files saved in:")
print(output_folder)
```

```python
cleaned_folder = "/content/drive/MyDrive/RAW/Cleaned"
standardized_folder = "/content/drive/MyDrive/RAW/Standardized"
os.makedirs(standardized_folder, exist_ok=True)


# ------------------------------------------------------------
# 2. LOAD CLEANED DATASETS
# ------------------------------------------------------------

patients = pd.read_csv(
    os.path.join(cleaned_folder, "patients_cleaned.csv"),
    low_memory=False
)

readmission = pd.read_csv(
    os.path.join(cleaned_folder, "readmission_cleaned.csv"),
    low_memory=False
)

inpatient = pd.read_csv(
    os.path.join(cleaned_folder, "inpatient_discharges_cleaned.csv"),
    low_memory=False
)


# ============================================================
# 3. PATIENT ID STANDARDIZATION
# ============================================================

# patients.csv already has IDs like PAT-09484753
patients["patient_id"] = (
    patients["patient_id"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Readmission dataset has numeric patient IDs
readmission["patient_id"] = (
    pd.to_numeric(
        readmission["patient_id"],
        errors="coerce"
    )
    .astype("Int64")
)

# Create a consistent text ID format
readmission["patient_id"] = (
    "PAT-" +
    readmission["patient_id"].astype(str)
)


# ============================================================
# 4. HOSPITAL ID STANDARDIZATION
# ============================================================

readmission["hospital_id"] = (
    pd.to_numeric(
        readmission["hospital_id"],
        errors="coerce"
    )
    .astype("Int64")
)

# Convert hospital ID to consistent text format
readmission["hospital_id"] = (
    "HOSP-" +
    readmission["hospital_id"].astype(str)
)


# ============================================================
# 5. DOCTOR ID STANDARDIZATION
# ============================================================

readmission["doctor_id"] = (
    pd.to_numeric(
        readmission["doctor_id"],
        errors="coerce"
    )
    .astype("Int64")
)

# Convert doctor ID to consistent text format
readmission["doctor_id"] = (
    "DOC-" +
    readmission["doctor_id"].astype(str)
)


# ============================================================
# 6. INPATIENT FACILITY ID STANDARDIZATION
# ============================================================

# Permanent Facility ID
inpatient["permanent_facility_id"] = pd.to_numeric(
    inpatient["permanent_facility_id"],
    errors="coerce"
)

inpatient["permanent_facility_id"] = (
    "FAC-" +
    inpatient["permanent_facility_id"]
    .astype("Int64")
    .astype(str)
)

# Operating Certificate Number
inpatient["operating_certificate_number"] = pd.to_numeric(
    inpatient["operating_certificate_number"],
    errors="coerce"
)

inpatient["operating_certificate_number"] = (
    "CERT-" +
    inpatient["operating_certificate_number"]
    .astype("Int64")
    .astype(str)
)


# ============================================================
# 7. VERIFY ID STANDARDIZATION
# ============================================================

print("=" * 70)
print("ID STANDARDIZATION RESULTS")
print("=" * 70)

print("\nPatients - Patient IDs:")
print(patients["patient_id"].head())

print("\nReadmission - Patient IDs:")
print(readmission["patient_id"].head())

print("\nReadmission - Hospital IDs:")
print(readmission["hospital_id"].head())

print("\nReadmission - Doctor IDs:")
print(readmission["doctor_id"].head())

print("\nInpatient - Facility IDs:")
print(inpatient["permanent_facility_id"].head())

print("\nInpatient - Certificate IDs:")
print(inpatient["operating_certificate_number"].head())


# ============================================================
# 8. SAVE STANDARDIZED DATASETS
# ============================================================

patients.to_csv(
    os.path.join(
        standardized_folder,
        "patients_standardized.csv"
    ),
    index=False
)

readmission.to_csv(
    os.path.join(
        standardized_folder,
        "readmission_standardized.csv"
    ),
    index=False
)

inpatient.to_csv(
    os.path.join(
        standardized_folder,
        "inpatient_discharges_standardized.csv"
    ),
    index=False
)


# ============================================================
# 9. FINAL CHECK
# ============================================================

print("\n" + "=" * 70)
print("ID STANDARDIZATION COMPLETED")
print("=" * 70)

print("\nPatients:", patients.shape)
print("Readmission:", readmission.shape)
print("Inpatient Discharges:", inpatient.shape)

print("\nStandardized files saved to:")
print(standardized_folder)
```
