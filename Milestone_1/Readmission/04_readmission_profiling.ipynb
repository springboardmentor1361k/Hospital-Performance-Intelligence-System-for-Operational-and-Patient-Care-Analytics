## Readmission Dataset – Profiling, Cleaning & Validation Summary

### Dataset Overview
- Dataset contains 10,000 rows and 26 columns.
- No missing values were found.
- No completely duplicate records were found.
- Admission_date, patient_checkin_date, and patient_checkout_date were converted to datetime format.

### Data Cleaning
- Hospital ward values were standardized by changing "ER" to "Emergency Department".
- Patient age was checked; values range from 0 to 95.
- Patient length of stay ranges from 1 to 30 days.
- Readmission contains valid binary values: 0 and 1.
- Negative hospital bed, occupied bed, hospital ID, and doctor ID values were not found.
- Patient gender and discharge status values were checked.

### Data Consistency Issues Identified

1. **Hospital Name vs Hospital ID**
   - Only 1 unique hospital name was found.
   - The hospital name "The Johns Hopkins Hospital" is associated with 6,074 different hospital IDs.
   - Original hospital IDs were preserved because the correct mapping could not be determined from the source data.

2. **Checkout Date Earlier Than Check-in Date**
   - 4,750 records have patient_checkout_date earlier than patient_checkin_date.
   - Original date values were preserved because the correct dates could not be determined reliably.

3. **Length of Stay Mismatch**
   - 9,822 records have a mismatch between the calculated date difference and patient_length_of_stay.
   - LOS values were not modified because the check-in and checkout dates themselves contain inconsistencies.

4. **Admission Date Later Than Check-in Date**
   - 87 records have Admission_date later than patient_checkin_date.
   - Original values were preserved because the correct date relationship could not be determined from the available source information.

5. **Doctor ID Mapped to Multiple Doctor Names**
   - 2,755 doctor IDs are associated with more than one doctor name.
   - Some IDs are also associated with different doctor specialties.
   - Example: Doctor ID 1002 is associated with three different doctor names and specialties.
   - Original doctor IDs and names were preserved because the correct mapping could not be determined reliably.

### Validation Decision
- The identified consistency issues were verified using Python checks.
- No values were arbitrarily changed or deleted.
- The original source values were preserved where reliable correction evidence was unavailable.
- The identified issues are documented for further validation with the source data/team.

### Output
- Final cleaned dataset contains 10,000 rows and 26 columns.
- Cleaned dataset saved as: `data/processed/readmission_cleaned.csv`
