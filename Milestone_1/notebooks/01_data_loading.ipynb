```python
datasets = {
    "Hospital Overview": hospital_overview,
    "Patient Flow": patient_flow,
    "Department Analytics": department_analytics,
    "Resource Utilization": resource_utilization
}

for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Missing values:", df.isnull().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())
```

```python
profiling_summary = []

for name, df in datasets.items():
    profiling_summary.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

profiling_report = pd.DataFrame(profiling_summary)

profiling_report
```

```python
profiling_report.to_csv(
    "../data/processed/profiling_report.csv",
    index=False
)

print("Profiling report saved successfully.")
```
