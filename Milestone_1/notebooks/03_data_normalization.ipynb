```python
import pandas as pd
import os
```

```python
hospital_clean = pd.read_csv("../data/processed/hospital_overview_clean.csv")
patient_flow_clean = pd.read_csv("../data/processed/patient_flow_clean.csv")
department_clean = pd.read_csv("../data/processed/department_analytics_clean.csv")
resource_clean = pd.read_csv("../data/processed/resource_utilization_clean.csv")
```

```python
print("Hospital Overview:", hospital_clean.shape)
print("Patient Flow:", patient_flow_clean.shape)
print("Department Analytics:", department_clean.shape)
print("Resource Utilization:", resource_clean.shape)
```

```python
hospital_clean["hospital_id"] = hospital_clean["hospital_id"].astype(str).str.strip().str.upper()
patient_flow_clean["hospital_id"] = patient_flow_clean["hospital_id"].astype(str).str.strip().str.upper()
department_clean["hospital_id"] = department_clean["hospital_id"].astype(str).str.strip().str.upper()
resource_clean["hospital_id"] = resource_clean["hospital_id"].astype(str).str.strip().str.upper()
```

```python
print("Hospital Overview:")
print(sorted(hospital_clean["hospital_id"].unique()))

print("\nPatient Flow:")
print(sorted(patient_flow_clean["hospital_id"].unique()))

print("\nDepartment Analytics:")
print(sorted(department_clean["hospital_id"].unique()))

print("\nResource Utilization:")
print(sorted(resource_clean["hospital_id"].unique()))
```

```python
hospital_clean["department_id"] = hospital_clean["department_id"].astype(str).str.strip().str.upper()
patient_flow_clean["current_department_id"] = patient_flow_clean["current_department_id"].astype(str).str.strip().str.upper()
patient_flow_clean["from_department_id"] = patient_flow_clean["from_department_id"].astype(str).str.strip().str.upper()
department_clean["department_id"] = department_clean["department_id"].astype(str).str.strip().str.upper()
resource_clean["department_id"] = resource_clean["department_id"].astype(str).str.strip().str.upper()
```

```python
print("Hospital Overview:")
print(sorted(hospital_clean["department_id"].unique()))

print("\nPatient Flow - Current Department:")
print(sorted(patient_flow_clean["current_department_id"].unique()))

print("\nDepartment Analytics:")
print(sorted(department_clean["department_id"].unique()))

print("\nResource Utilization:")
print(sorted(resource_clean["department_id"].unique()))
```

```python
hospital_clean["department_name"] = (
    hospital_clean["department_name"]
    .astype(str)
    .str.strip()
    .str.lower()
)

patient_flow_clean["current_department_name"] = (
    patient_flow_clean["current_department_name"]
    .astype(str)
    .str.strip()
    .str.lower()
)

patient_flow_clean["from_department_name"] = (
    patient_flow_clean["from_department_name"]
    .astype(str)
    .str.strip()
    .str.lower()
)

department_clean["department_name"] = (
    department_clean["department_name"]
    .astype(str)
    .str.strip()
    .str.lower()
)

resource_clean["department_name"] = (
    resource_clean["department_name"]
    .astype(str)
    .str.strip()
    .str.lower()
)
```

```python
print("Hospital Overview:")
print(sorted(hospital_clean["department_name"].unique()))

print("\nPatient Flow:")
print(sorted(patient_flow_clean["current_department_name"].unique()))

print("\nDepartment Analytics:")
print(sorted(department_clean["department_name"].unique()))

print("\nResource Utilization:")
print(sorted(resource_clean["department_name"].unique()))
```

```python
print("Hospital Overview admission IDs:")
print(hospital_clean["admission_id"].head())

print("\nPatient Flow admission IDs:")
print(patient_flow_clean["admission_id"].head())
```

```python
hospital_admissions = set(hospital_clean["admission_id"])

flow_admissions = set(patient_flow_clean["admission_id"])

missing_admissions = flow_admissions - hospital_admissions

print("Admission IDs in Patient Flow not found in Hospital Overview:")
print(missing_admissions)

print("\nNumber of unmatched admission IDs:", len(missing_admissions))
```

```python
hospital_patients = set(hospital_clean["patient_id"])
flow_patients = set(patient_flow_clean["patient_id"])

missing_patients = flow_patients - hospital_patients

print("Patient IDs in Patient Flow not found in Hospital Overview:")
print(missing_patients)

print("\nNumber of unmatched patient IDs:", len(missing_patients))
```

```python
hospital_clean["admission_date"] = pd.to_datetime(
    hospital_clean["admission_date"],
    errors="coerce"
)

hospital_clean["discharge_date"] = pd.to_datetime(
    hospital_clean["discharge_date"],
    errors="coerce"
)

patient_flow_clean["movement_date"] = pd.to_datetime(
    patient_flow_clean["movement_date"],
    errors="coerce"
)

department_clean["date"] = pd.to_datetime(
    department_clean["date"],
    errors="coerce"
)

resource_clean["date"] = pd.to_datetime(
    resource_clean["date"],
    errors="coerce"
)

resource_clean["last_maintenance_date"] = pd.to_datetime(
    resource_clean["last_maintenance_date"],
    errors="coerce"
)

print("Date columns converted successfully.")
```

```python
print("Hospital Overview admission date type:")
print(hospital_clean["admission_date"].dtype)

print("\nHospital Overview discharge date type:")
print(hospital_clean["discharge_date"].dtype)

print("\nPatient Flow movement date type:")
print(patient_flow_clean["movement_date"].dtype)

print("\nDepartment date type:")
print(department_clean["date"].dtype)

print("\nResource date type:")
print(resource_clean["date"].dtype)
```

```python
hospital_clean["date"] = hospital_clean["admission_date"].dt.date
patient_flow_clean["date"] = patient_flow_clean["movement_date"].dt.date
```

```python
hospital_clean["date"] = pd.to_datetime(hospital_clean["date"])
patient_flow_clean["date"] = pd.to_datetime(patient_flow_clean["date"])
```

```python
print("Admission types:")
print(hospital_clean["admission_type"].value_counts())

print("\nAdmission sources:")
print(hospital_clean["admission_source"].value_counts())

print("\nPayment status:")
print(hospital_clean["payment_status"].value_counts())

print("\nDischarge status:")
print(hospital_clean["discharge_status"].value_counts())
```

```python
print("Movement types:")
print(patient_flow_clean["movement_type"].value_counts())
```

```python
print("Resource types:")
print(resource_clean["resource_type"].value_counts())
```
