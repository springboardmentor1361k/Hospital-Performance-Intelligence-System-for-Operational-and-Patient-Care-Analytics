```python
import pandas as pd
import os
```

```python
hospital_overview = pd.read_csv(
    "../data/raw/hospital_overview_dataset.csv"
)

patient_flow = pd.read_csv(
    "../data/raw/patient_flow_dataset.csv"
)

department_analytics = pd.read_csv(
    "../data/raw/department_analytics_dataset.csv"
)

resource_utilization = pd.read_csv(
    "../data/raw/resource_utilization_dataset.csv"
)
```

```python
hospital_clean = hospital_overview.copy()
patient_flow_clean = patient_flow.copy()
department_clean = department_analytics.copy()
resource_clean = resource_utilization.copy()
```

```python
print("Hospital Overview:")
print(hospital_clean.isnull().sum().sum())

print("\nPatient Flow:")
print(patient_flow_clean.isnull().sum().sum())

print("\nDepartment Analytics:")
print(department_clean.isnull().sum().sum())

print("\nResource Utilization:")
print(resource_clean.isnull().sum().sum())
```

```python
print("Hospital Overview:", hospital_clean.duplicated().sum())
print("Patient Flow:", patient_flow_clean.duplicated().sum())
print("Department Analytics:", department_clean.duplicated().sum())
print("Resource Utilization:", resource_clean.duplicated().sum())
```

```python
def clean_text_columns(df):
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].str.strip()
    return df
```

```python
hospital_clean = clean_text_columns(hospital_clean)
patient_flow_clean = clean_text_columns(patient_flow_clean)
department_clean = clean_text_columns(department_clean)
resource_clean = clean_text_columns(resource_clean)
```

```python
print("Hospital Overview Departments:")
print(sorted(hospital_clean["department_name"].unique()))

print("\nDepartment Analytics Departments:")
print(sorted(department_clean["department_name"].unique()))

print("\nPatient Flow Departments:")
print(sorted(patient_flow_clean["current_department_name"].unique()))
```

```python
print("Hospital Overview:")
print(sorted(hospital_clean["hospital_name"].unique()))

print("\nDepartment Analytics:")
print(sorted(department_clean["hospital_name"].unique()))

print("\nResource Utilization:")
print(sorted(resource_clean["hospital_name"].unique()))
```

```python
print("Hospital IDs:")
print(hospital_clean["hospital_id"].unique())

print("\nDepartment IDs:")
print(hospital_clean["department_id"].unique())

print("\nPatient IDs:")
print(hospital_clean["patient_id"].head())

print("\nAdmission IDs:")
print(hospital_clean["admission_id"].head())
```

```python
hospital_clean["admission_date"] = pd.to_datetime(
    hospital_clean["admission_date"],
    errors="coerce"
)

hospital_clean["discharge_date"] = pd.to_datetime(
    hospital_clean["discharge_date"],
    errors="coerce"
)
```

```python
patient_flow_clean["movement_datetime"] = pd.to_datetime(
    patient_flow_clean["movement_datetime"],
    errors="coerce"
)

patient_flow_clean["movement_date"] = pd.to_datetime(
    patient_flow_clean["movement_date"],
    errors="coerce"
)
```

```python
department_clean["date"] = pd.to_datetime(
    department_clean["date"],
    errors="coerce"
)

resource_clean["date"] = pd.to_datetime(
    resource_clean["date"],
    errors="coerce"
)

resource_clean["last_maintenance_date"] = pd.to_datetime(
    resource_clean["last_maintenance_date"],
    errors="coerce"
)
```

```python
print("Hospital admission date errors:",
      hospital_clean["admission_date"].isna().sum())

print("Hospital discharge date errors:",
      hospital_clean["discharge_date"].isna().sum())

print("Patient movement date errors:",
      patient_flow_clean["movement_date"].isna().sum())

print("Department date errors:",
      department_clean["date"].isna().sum())

print("Resource date errors:",
      resource_clean["date"].isna().sum())
```

```python
print("Patient age:")
print(hospital_clean["patient_age"].describe())
```

```python
print("Patient satisfaction:")
print(hospital_clean["patient_satisfaction_score"].describe())
```

```python
print("Length of stay:")
print(hospital_clean["length_of_stay_days"].describe())
```

```python
print("Occupancy rate:")
print(department_clean["bed_occupancy_rate_pct"].describe())
```

```python
print("Resource utilization rate:")
print(resource_clean["utilization_rate_pct"].describe())
```

```python
print("Occupancy values outside 0-100:")
print(
    department_clean[
        (department_clean["bed_occupancy_rate_pct"] < 0) |
        (department_clean["bed_occupancy_rate_pct"] > 100)
    ]
)
```

```python
print("Resource utilization values outside 0-100:")
print(
    resource_clean[
        (resource_clean["utilization_rate_pct"] < 0) |
        (resource_clean["utilization_rate_pct"] > 100)
    ]
)
```

```python
invalid_dates = hospital_clean[
    hospital_clean["discharge_date"] < hospital_clean["admission_date"]
]

print("Invalid admission/discharge dates:", len(invalid_dates))
```

```python
invalid_beds = department_clean[
    department_clean["occupied_beds_count"] >
    department_clean["total_beds"]
]

print("Invalid bed records:", len(invalid_beds))
```

```python
invalid_resources = resource_clean[
    (
        resource_clean["units_in_use"] +
        resource_clean["units_under_maintenance"]
    )
    >
    resource_clean["total_units_available"]
]

print("Invalid resource records:", len(invalid_resources))
```

```python
print("CLEANING SUMMARY")
print("=" * 50)

print("Hospital Overview")
print("Rows:", len(hospital_clean))
print("Missing:", hospital_clean.isnull().sum().sum())
print("Duplicates:", hospital_clean.duplicated().sum())

print("\nPatient Flow")
print("Rows:", len(patient_flow_clean))
print("Missing:", patient_flow_clean.isnull().sum().sum())
print("Duplicates:", patient_flow_clean.duplicated().sum())

print("\nDepartment Analytics")
print("Rows:", len(department_clean))
print("Missing:", department_clean.isnull().sum().sum())
print("Duplicates:", department_clean.duplicated().sum())

print("\nResource Utilization")
print("Rows:", len(resource_clean))
print("Missing:", resource_clean.isnull().sum().sum())
print("Duplicates:", resource_clean.duplicated().sum())
```

```python
hospital_clean.to_csv(
    "../data/processed/hospital_overview_clean.csv",
    index=False
)

patient_flow_clean.to_csv(
    "../data/processed/patient_flow_clean.csv",
    index=False
)

department_clean.to_csv(
    "../data/processed/department_analytics_clean.csv",
    index=False
)

resource_clean.to_csv(
    "../data/processed/resource_utilization_clean.csv",
    index=False
)

print("All cleaned datasets saved successfully.")
```
