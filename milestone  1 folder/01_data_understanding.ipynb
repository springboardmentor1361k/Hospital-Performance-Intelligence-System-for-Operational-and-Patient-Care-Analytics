```python
import pandas as pd
import os

files = os.listdir()

profile = []

for file in files:

    try:
        
        if file.lower().endswith(".csv"):
            df = pd.read_csv(file)

        
        elif file.lower().endswith(".xlsx"):
            df = pd.read_excel(file, engine="openpyxl")

        elif file.lower().endswith(".xls"):
            df = pd.read_excel(file, engine="xlrd")

        else:
            continue

        profile.append({
            "File": file,
            "Rows": df.shape[0],
            "Columns": df.shape[1],
            "Missing Values": df.isnull().sum().sum(),
            "Duplicate Rows": df.duplicated().sum()
        })

    except Exception as e:
        print("ERROR:", file)
        print(e)

profile_df = pd.DataFrame(profile)

profile_df
```

```python
# ==============================
# HOSPITAL OVERVIEW TABLE
# ==============================

import pandas as pd

# Load required datasets
patient = pd.read_csv("patient.csv")
admission = pd.read_csv("admission.csv")
billing = pd.read_csv("billing.csv")
bed = pd.read_csv("bed.csv")
department = pd.read_csv("department.csv")
doctor = pd.read_csv("doctor.csv")
employee = pd.read_csv("employee.csv")

# Calculate Average Length of Stay
admission["admission_date"] = pd.to_datetime(admission["admission_date"])
admission["discharge_date"] = pd.to_datetime(admission["discharge_date"])

admission["length_of_stay"] = (
    admission["discharge_date"] - admission["admission_date"]
).dt.days

# Create Hospital Overview
hospital_overview = pd.DataFrame({
    "Metric": [
        "Total Patients",
        "Total Admissions",
        "Total Bills",
        "Total Beds",
        "Total Departments",
        "Total Doctors",
        "Total Employees",
        "Total Revenue",
        "Average Length of Stay"
    ],

    "Value": [
        len(patient),
        len(admission),
        len(billing),
        len(bed),
        len(department),
        len(doctor),
        len(employee),
        billing["total_amount"].sum(),
        admission["length_of_stay"].mean()
    ]
})

hospital_overview
```

```python
# ==========================================
# STEP 3 - PATIENT FLOW TABLE
# ==========================================

# Admission dates ko datetime me convert karna
admission["admission_date"] = pd.to_datetime(admission["admission_date"])
admission["discharge_date"] = pd.to_datetime(admission["discharge_date"])

# Length of Stay calculate karna
admission["length_of_stay"] = (
    admission["discharge_date"] - admission["admission_date"]
).dt.days

# Patient Flow table
patient_flow = admission[
    [
        "admission_id",
        "patient_id",
        "admission_date",
        "discharge_date",
        "admission_type",
        "admission_status",
        "department_id",
        "ward_id",
        "bed_id",
        "disease_id",
        "length_of_stay"
    ]
].copy()

# First 10 records
patient_flow.head(10)
```

```python
# ==========================================
# STEP 4 - DEPARTMENT ANALYTICS TABLE
# ==========================================

# Department-wise admission analysis
department_analytics = admission.groupby("department_id").agg(
    total_admissions=("admission_id", "count"),
    average_length_of_stay=("length_of_stay", "mean")
).reset_index()

# Department-wise revenue
department_revenue = billing.merge(
    admission[["admission_id", "department_id"]],
    on="admission_id",
    how="left"
)

department_revenue = department_revenue.groupby("department_id").agg(
    total_revenue=("total_amount", "sum")
).reset_index()

# Combine admission + revenue data
department_analytics = department_analytics.merge(
    department_revenue,
    on="department_id",
    how="left"
)

# Add department names
department_analytics = department_analytics.merge(
    department[["department_id", "department_name"]],
    on="department_id",
    how="left"
)

# Arrange columns
department_analytics = department_analytics[
    [
        "department_id",
        "department_name",
        "total_admissions",
        "average_length_of_stay",
        "total_revenue"
    ]
]

# Round average stay
department_analytics["average_length_of_stay"] = (
    department_analytics["average_length_of_stay"].round(2)
)

department_analytics
```

```python
# ==========================================
# STEP 5 - RESOURCE UTILIZATION TABLE
# ==========================================

# Ward-wise occupied bed days
ward_usage = admission.groupby("ward_id").agg(
    total_admissions=("admission_id", "count"),
    occupied_bed_days=("length_of_stay", "sum")
).reset_index()

# Add ward information
resource_utilization = ward_usage.merge(
    ward[["ward_id", "ward_name", "ward_type", "total_beds", "department_id"]],
    on="ward_id",
    how="left"
)

# Calculate total number of days in the dataset period
start_date = admission["admission_date"].min()
end_date = admission["discharge_date"].max()

period_days = (end_date - start_date).days + 1

# Available bed days
resource_utilization["available_bed_days"] = (
    resource_utilization["total_beds"] * period_days
)

# Bed utilization percentage
resource_utilization["bed_utilization_percent"] = (
    resource_utilization["occupied_bed_days"]
    / resource_utilization["available_bed_days"]
) * 100

# Round percentage
resource_utilization["bed_utilization_percent"] = (
    resource_utilization["bed_utilization_percent"].round(2)
)

# Final columns
resource_utilization = resource_utilization[
    [
        "ward_id",
        "ward_name",
        "ward_type",
        "department_id",
        "total_beds",
        "total_admissions",
        "occupied_bed_days",
        "available_bed_days",
        "bed_utilization_percent"
    ]
]

resource_utilization
```

```python
# ==========================================
# BED MANAGEMENT - WEEKLY RESOURCE TABLE
# ==========================================

bed_resource_weekly = (
    bed_management
    .groupby(["Service", "Week"])
    .agg(
        available_beds=("Available Beds (Wk)", "max"),
        patients_requested=("Patients Requested (Wk)", "sum"),
        patients_admitted=("Patients Admitted (Wk)", "sum"),
        patients_refused=("Patients Refused (Wk)", "sum")
    )
    .reset_index()
)

bed_resource_weekly["admission_rate"] = (
    bed_resource_weekly["patients_admitted"]
    / bed_resource_weekly["patients_requested"]
    * 100
).round(2)

bed_resource_weekly["refusal_rate"] = (
    bed_resource_weekly["patients_refused"]
    / bed_resource_weekly["patients_requested"]
    * 100
).round(2)

display(bed_resource_weekly.head(20))
```

```python
# ==========================================
# ADD STAFF & RESOURCE METRICS
# ==========================================

staff_resource = (
    bed_management
    .groupby(["Service", "Week"])
    .agg(
        total_staff=("Dept Total Staff", "max"),
        doctor_count=("Dept Doctor Count", "max"),
        nurse_count=("Dept Nurse Count", "max"),
        nursing_assistant_count=("Dept Nursing Assistant Count", "max"),
        avg_satisfaction=("Avg Satisfaction (Wk)", "max"),
        staff_morale=("Staff Morale (Wk)", "max")
    )
    .reset_index()
)

display(staff_resource.head(20))
```

```python
# ==========================================
# FINAL BED MANAGEMENT RESOURCE TABLE
# ==========================================

bed_management_resource = bed_resource_weekly.merge(
    staff_resource,
    on=["Service", "Week"],
    how="left"
)

print("===== BED MANAGEMENT RESOURCE TABLE =====")
print("Rows:", len(bed_management_resource))
print("Columns:", len(bed_management_resource.columns))

display(bed_management_resource.head(20))
```

```python
# ==========================================
# FINAL TABLE 2 - PATIENT & CLINICAL ANALYTICS
# ==========================================

# 1. Patient-wise admission summary
patient_admission = (
    admission
    .groupby("patient_id")
    .agg(
        total_admissions=("admission_id", "nunique"),
        average_length_of_stay=("length_of_stay", "mean"),
        total_length_of_stay=("length_of_stay", "sum")
    )
    .reset_index()
)


# 2. Patient-wise diagnostic test count
patient_diagnostic_count = (
    patient_diagnostic
    .groupby("admission_id")
    .size()
    .reset_index(name="diagnostic_tests")
)

# Map admission_id -> patient_id
diagnostic_patient = patient_diagnostic_count.merge(
    admission[["admission_id", "patient_id"]],
    on="admission_id",
    how="left"
)

diagnostic_patient = (
    diagnostic_patient
    .groupby("patient_id")
    ["diagnostic_tests"]
    .sum()
    .reset_index()
)


# 3. Patient-wise prescription count
patient_prescription_count = (
    prescription
    .groupby("admission_id")
    .size()
    .reset_index(name="prescription_count")
)

# Map admission_id -> patient_id
prescription_patient = patient_prescription_count.merge(
    admission[["admission_id", "patient_id"]],
    on="admission_id",
    how="left"
)

prescription_patient = (
    prescription_patient
    .groupby("patient_id")
    ["prescription_count"]
    .sum()
    .reset_index()
)


# 4. Patient-wise disease count
patient_disease = (
    admission
    .groupby("patient_id")
    ["disease_id"]
    .nunique()
    .reset_index(name="unique_diseases")
)


# 5. Readmission information
readmission_patient = (
    readmission
    .groupby("patient_id")
    .agg(
        readmission_records=("readmission", "count"),
        readmitted=("readmission", "max")
    )
    .reset_index()
)


# 6. Combine everything
patient_clinical = patient[
    ["patient_id", "gender", "blood_group", "city"]
].copy()

patient_clinical = patient_clinical.merge(
    patient_admission,
    on="patient_id",
    how="left"
)

patient_clinical = patient_clinical.merge(
    diagnostic_patient,
    on="patient_id",
    how="left"
)

patient_clinical = patient_clinical.merge(
    prescription_patient,
    on="patient_id",
    how="left"
)

patient_clinical = patient_clinical.merge(
    patient_disease,
    on="patient_id",
    how="left"
)

patient_clinical = patient_clinical.merge(
    readmission_patient,
    on="patient_id",
    how="left"
)


# 7. Fill patients with no diagnostic/prescription/readmission record
patient_clinical[
    ["diagnostic_tests",
     "prescription_count",
     "unique_diseases",
     "readmission_records",
     "readmitted"]
] = patient_clinical[
    ["diagnostic_tests",
     "prescription_count",
     "unique_diseases",
     "readmission_records",
     "readmitted"]
].fillna(0)


# 8. Round average LOS
patient_clinical["average_length_of_stay"] = (
    patient_clinical["average_length_of_stay"].round(2)
)


print("===== FINAL TABLE 2 - PATIENT & CLINICAL ANALYTICS =====")

print("Rows:", len(patient_clinical))
print("Columns:", len(patient_clinical.columns))

display(patient_clinical.head(10))
```

```python
# ==========================================
# FINAL TABLE 3 - DEPARTMENT & RESOURCE ANALYTICS
# ==========================================

# 1. Admissions per department
department_admissions = (
    admission
    .groupby("department_id")
    .agg(
        total_admissions=("admission_id", "nunique"),
        average_length_of_stay=("length_of_stay", "mean")
    )
    .reset_index()
)

# 2. Wards and beds per department
department_wards = (
    ward
    .groupby("department_id")
    .agg(
        total_wards=("ward_id", "nunique"),
        total_ward_beds=("total_beds", "sum")
    )
    .reset_index()
)

# 3. Employees per department
department_staff = (
    employee
    .groupby("department_id")
    .agg(
        total_staff=("employee_id", "nunique")
    )
    .reset_index()
)

# 4. Doctors per department
# Doctor -> Employee -> Department
doctor_department = doctor.merge(
    employee[["employee_id", "department_id"]],
    on="employee_id",
    how="left"
)

department_doctors = (
    doctor_department
    .groupby("department_id")
    .agg(
        total_doctors=("doctor_id", "nunique")
    )
    .reset_index()
)

# 5. Bed status per department
bed_department = bed.merge(
    ward[["ward_id", "department_id"]],
    on="ward_id",
    how="left"
)

department_bed_status = (
    bed_department
    .groupby("department_id")
    .agg(
        total_bed_records=("bed_id", "nunique"),
        occupied_beds=("bed_status", lambda x:
                       (x.astype(str).str.lower() == "occupied").sum()),
        available_beds=("bed_status", lambda x:
                        (x.astype(str).str.lower() == "available").sum())
    )
    .reset_index()
)

# 6. Combine all department-level information
department_resource = department[
    ["department_id", "department_name", "department_type",
     "floor_number", "status"]
].copy()

department_resource = department_resource.merge(
    department_admissions,
    on="department_id",
    how="left"
)

department_resource = department_resource.merge(
    department_wards,
    on="department_id",
    how="left"
)

department_resource = department_resource.merge(
    department_staff,
    on="department_id",
    how="left"
)

department_resource = department_resource.merge(
    department_doctors,
    on="department_id",
    how="left"
)

department_resource = department_resource.merge(
    department_bed_status,
    on="department_id",
    how="left"
)

# 7. Fill missing values
numeric_columns = [
    "total_admissions",
    "average_length_of_stay",
    "total_wards",
    "total_ward_beds",
    "total_staff",
    "total_doctors",
    "total_bed_records",
    "occupied_beds",
    "available_beds"
]

department_resource[numeric_columns] = (
    department_resource[numeric_columns].fillna(0)
)

# 8. Round average LOS
department_resource["average_length_of_stay"] = (
    department_resource["average_length_of_stay"].round(2)
)

print("===== FINAL TABLE 3 - DEPARTMENT & RESOURCE ANALYTICS =====")

print("Rows:", len(department_resource))
print("Columns:", len(department_resource.columns))

display(department_resource)
```

```python
# ==========================================
# FINAL TABLE 1 - VALIDATION
# ==========================================

print("===== HOSPITAL OVERVIEW VALIDATION =====")

print("Original Admission Rows:", len(admission))
print("Final Table Rows:", len(hospital_overview))

print("\nOriginal Unique Admissions:",
      admission["admission_id"].nunique())

print("Final Unique Admissions:",
      hospital_overview["admission_id"].nunique())

print("\nDuplicate Admission IDs in Final Table:",
      hospital_overview["admission_id"].duplicated().sum())

print("\nMissing Admission IDs:",
      hospital_overview["admission_id"].isna().sum())
```

```python
# ==========================================
# FINAL TABLE 2 - PATIENT FLOW
# STEP 1: CHECK READMISSION RELATIONSHIP
# ==========================================

print("===== READMISSION DATA =====")

print("Readmission Rows:", len(readmission))

print("Unique Patient IDs:",
      readmission["patient_id"].nunique())

print("Duplicate Patient IDs:",
      readmission["patient_id"].duplicated().sum())

print("\nAdmission Patient IDs:",
      admission["patient_id"].nunique())

print("Readmission columns:")
print(readmission.columns.tolist())
```

```python
# ==========================================
# FINAL TABLE 2 - PATIENT FLOW
# STEP 2: CREATE PATIENT-LEVEL READMISSION SUMMARY
# ==========================================

readmission_patient_summary = (
    readmission
    .groupby("patient_id")
    .agg(
        readmission_records=("patient_id", "size"),
        total_readmissions=("readmission", "sum"),
        average_wait_time=("patient_waittime", "mean"),
        average_satisfaction=("patient_sat_score", "mean"),
        average_length_of_stay=("patient_length_of_stay", "mean")
    )
    .reset_index()
)

print("===== READMISSION PATIENT SUMMARY =====")
print("Rows:", readmission_patient_summary.shape[0])
print("Columns:", readmission_patient_summary.shape[1])

print("\nColumns:")
print(readmission_patient_summary.columns.tolist())

print("\nFirst 10 rows:")
display(readmission_patient_summary.head(10))
```

```python
# ==========================================
# FINAL TABLE 2 - PATIENT FLOW
# STEP 3: CREATE BASE TABLE
# ==========================================

patient_flow = admission.copy()

# Patient information
patient_flow = patient_flow.merge(
    patient[
        [
            "patient_id",
            "gender",
            "date_of_birth",
            "blood_group",
            "city"
        ]
    ],
    on="patient_id",
    how="left",
    validate="many_to_one"
)

# Department information
patient_flow = patient_flow.merge(
    department[
        [
            "department_id",
            "department_name",
            "department_type"
        ]
    ],
    on="department_id",
    how="left",
    validate="many_to_one"
)

# Ward information
patient_flow = patient_flow.merge(
    ward[
        [
            "ward_id",
            "ward_name",
            "ward_type"
        ]
    ],
    on="ward_id",
    how="left",
    validate="many_to_one"
)

# Disease information
patient_flow = patient_flow.merge(
    disease[
        [
            "disease_id",
            "disease_name",
            "disease_category"
        ]
    ],
    on="disease_id",
    how="left",
    validate="many_to_one"
)

print("===== PATIENT FLOW BASE TABLE =====")
print("Rows:", patient_flow.shape[0])
print("Columns:", patient_flow.shape[1])

print("\nColumns:")
print(patient_flow.columns.tolist())

print("\nFirst 5 rows:")
display(patient_flow.head())
```

```python
# ==========================================
# FINAL TABLE 2 - PATIENT FLOW
# STEP 4: ADD READMISSION SUMMARY
# ==========================================

patient_flow = patient_flow.merge(
    readmission_patient_summary[
        [
            "patient_id",
            "readmission_records",
            "total_readmissions",
            "average_wait_time",
            "average_satisfaction",
            "average_length_of_stay"
        ]
    ],
    on="patient_id",
    how="left",
    validate="many_to_one"
)

print("===== PATIENT FLOW WITH READMISSION =====")
print("Rows:", patient_flow.shape[0])
print("Columns:", patient_flow.shape[1])

print("\nColumns:")
print(patient_flow.columns.tolist())

print("\nFirst 5 rows:")
display(patient_flow.head())
```

```python
# ==========================================
# FINAL TABLE 2 - PATIENT FLOW VALIDATION
# ==========================================

print("===== PATIENT FLOW VALIDATION =====")

print("Original Admission Rows:", len(admission))
print("Final Patient Flow Rows:", len(patient_flow))

print("\nOriginal Unique Admissions:",
      admission["admission_id"].nunique())

print("Final Unique Admissions:",
      patient_flow["admission_id"].nunique())

print("\nDuplicate Admission IDs:",
      patient_flow["admission_id"].duplicated().sum())

print("\nMissing Admission IDs:",
      patient_flow["admission_id"].isna().sum())

print("\nPatients with Readmission Data:",
      patient_flow["readmission_records"].notna().sum())

print("Patients without Readmission Data:",
      patient_flow["readmission_records"].isna().sum())
```

```python
# ==========================================
# FINAL TABLE 3 - DEPARTMENT ANALYTICS
# STEP 1: DEPARTMENT SUMMARY
# ==========================================

department_analytics = (
    patient_flow
    .groupby(
        [
            "department_id",
            "department_name",
            "department_type"
        ],
        as_index=False
    )
    .agg(
        total_admissions=("admission_id", "nunique"),
        unique_patients=("patient_id", "nunique"),
        average_length_of_stay=("length_of_stay", "mean")
    )
)

print("===== DEPARTMENT ANALYTICS BASE =====")
print("Rows:", department_analytics.shape[0])
print("Columns:", department_analytics.shape[1])

print("\nColumns:")
print(department_analytics.columns.tolist())

print("\nDepartment Summary:")
display(department_analytics)
```

```python
# ==========================================
# FINAL TABLE 3 - DEPARTMENT ANALYTICS
# STEP 2: ADD READMISSION METRICS
# ==========================================

readmission_dept = (
    patient_flow
    .groupby(
        [
            "department_id",
            "department_name"
        ],
        as_index=False
    )
    .agg(
        total_readmission_records=("total_readmissions", "sum"),
        patients_with_readmission_data=("readmission_records", "count")
    )
)

# Readmission rate
readmission_dept["readmission_rate"] = (
    readmission_dept["total_readmission_records"]
    / readmission_dept["patients_with_readmission_data"]
    * 100
)

print("===== DEPARTMENT READMISSION SUMMARY =====")
display(readmission_dept)
```

```python
# ==========================================
# FINAL TABLE 3 - DEPARTMENT ANALYTICS
# STEP 3: CORRECT READMISSION METRICS
# ==========================================

department_readmission = (
    readmission
    .groupby("department_referral", as_index=False)
    .agg(
        readmission_records=("readmission", "count"),
        total_readmitted=("readmission", "sum")
    )
)

department_readmission["readmission_rate"] = (
    department_readmission["total_readmitted"]
    / department_readmission["readmission_records"]
    * 100
)

print("===== CORRECT DEPARTMENT READMISSION =====")
display(department_readmission)
```

```python
# ==========================================
# FINAL TABLE 3 - DEPARTMENT ANALYTICS
# STEP 4: CHECK DEPARTMENT NAME MATCHING
# ==========================================

print("===== HMIS DEPARTMENTS =====")
print(department_analytics["department_name"].tolist())

print("\n===== READMISSION DEPARTMENTS =====")
print(department_readmission["department_referral"].tolist())

# Exact department name matches
matched_departments = set(
    department_analytics["department_name"]
) & set(
    department_readmission["department_referral"]
)

print("\n===== EXACT MATCHES =====")
print(matched_departments)

print("\nNumber of Exact Matches:",
      len(matched_departments))
```

```python
# ==========================================
# FINAL TABLE 3 - DEPARTMENT ANALYTICS
# STEP 5: ADD DEPARTMENT BED RESOURCES
# ==========================================

department_beds = (
    ward
    .groupby("department_id", as_index=False)
    .agg(
        total_beds=("total_beds", "sum"),
        total_wards=("ward_id", "nunique")
    )
)

print("===== DEPARTMENT BED RESOURCES =====")
display(department_beds)
```

```python
# ==========================================
# FINAL TABLE 3 - DEPARTMENT ANALYTICS
# STEP 6: MERGE BED RESOURCES
# ==========================================

department_analytics = department_analytics.merge(
    department_beds,
    on="department_id",
    how="left",
    validate="one_to_one"
)

print("===== DEPARTMENT ANALYTICS =====")
print("Rows:", department_analytics.shape[0])
print("Columns:", department_analytics.shape[1])

print("\nColumns:")
print(department_analytics.columns.tolist())

print("\nFinal Department Analytics:")
display(department_analytics)
```

```python
# ==========================================
# FINAL TABLE 3 - DEPARTMENT ANALYTICS
# STEP 7: DOCTOR & STAFF RESOURCE SUMMARY
# ==========================================

# Doctor -> Employee -> Department
doctor_department = doctor.merge(
    employee[
        ["employee_id", "department_id"]
    ],
    on="employee_id",
    how="left",
    validate="many_to_one"
)

# Department-wise doctor count
department_doctors = (
    doctor_department
    .groupby("department_id", as_index=False)
    .agg(
        total_doctors=("doctor_id", "nunique")
    )
)

# Department-wise employee/staff count
department_staff = (
    employee
    .groupby("department_id", as_index=False)
    .agg(
        total_staff=("employee_id", "nunique")
    )
)

print("===== DEPARTMENT DOCTOR RESOURCES =====")
display(department_doctors)

print("\n===== DEPARTMENT STAFF RESOURCES =====")
display(department_staff)
```

```python
# ==========================================
# FINAL TABLE 3 - DEPARTMENT ANALYTICS
# STEP 8: ADD DOCTOR & STAFF RESOURCES
# ==========================================

department_analytics = department_analytics.merge(
    department_doctors,
    on="department_id",
    how="left",
    validate="one_to_one"
)

department_analytics = department_analytics.merge(
    department_staff,
    on="department_id",
    how="left",
    validate="one_to_one"
)

print("===== DEPARTMENT ANALYTICS WITH RESOURCES =====")

print("Rows:", department_analytics.shape[0])
print("Columns:", department_analytics.shape[1])

print("\nColumns:")
print(department_analytics.columns.tolist())

print("\nFinal Table 3 Preview:")
display(department_analytics)
```
