#### **1.First Importing the cleaned Sets**

```python
#Here importent Libraries will be imported as needed.
import pandas as pd
```

```python
#Here all the four cleaned datasets will be readed into there respective variables as dataframes.
hospital_df=pd.read_csv("C:\\powerbipro\\Health_Domain_pro_Infosys\\MedTrack_DV\\Data\\Processed_data\\Hospital_cleaned.csv")
patient_df=pd.read_csv("C:\\powerbipro\\Health_Domain_pro_Infosys\\MedTrack_DV\\Data\\Processed_data\\Patient_cleaned.csv")
department_df=pd.read_csv("C:\\powerbipro\\Health_Domain_pro_Infosys\\MedTrack_DV\\Data\\Processed_data\\Department_cleaned.csv")
resource_df=pd.read_csv("C:\\powerbipro\\Health_Domain_pro_Infosys\\MedTrack_DV\\Data\\Processed_data\\Resources_cleaned.csv")
```

#### **2. Checking required keys exist**

```python
datasets = {
    'Hospital Overview': hospital_df,
    'Patient Flow': patient_df,
    'Department Analytics': department_df,
    'Resource Utilization': resource_df
}

for name, df in datasets.items():
    print(f"\n{name}")
    print(df.columns.tolist())
```

```python
linking_keys = [
    
    'Hospital_id',
    'Department_id',
    'Patient_id',
    'Admission_id',
    'Date',
    'Admission_type',
    'Movement_date'
]

for name, df in datasets.items():
    print(f"\n{name}")
    
    for key in linking_keys:
        print(f"{key}: {'Available' if key in df.columns else 'Missing'}")
```

#### **3. Validate data types**

```python
#Here we are checking whether the columns which we will be used to further joins among the tables have the same dtypes.
for name, df in datasets.items():
    print(f"\n{name}")
    
    for key in linking_keys:
        if key in df.columns:
            print(f"{key}: {df[key].dtype}")   #Here date column converted again into the object becuase of the default behavior of the pandas csv reading 
```

#### **4.Check missing linking IDs**

```python
#Here we checking whether any column keys does have any missing values.
for name, df in datasets.items():
    print(f"\n{name}")
    
    for key in linking_keys:
        if key in df.columns:
            print(f"{key}: {df[key].isna().sum()} missing")
```

Here we couldn't got any missing values by the key columns which we want to linking.    

so we couldn't face any problem while linking the datasets.

#### **5.Check uniqueness**


```python
#Here we checking whether any column keys does have any missing values.
for name, df in datasets.items():
    print(f"\n{name}")
    
    for key in linking_keys:
        if key in df.columns:
            print(f"{key}: {df[key].is_unique }")
```

Based on the grain the uniqueness was correctly maintained across the four datasets.

#### **7.Check Hospital + Department combinations**

```python
#Here we create a set dtype of combination by using the zip() ,it's returns the zip object return type
department_pairs = set(
    zip(
        department_df['Hospital_id'],
        department_df['Department_id']
    )
)

#Here we create a set dtype of the resource type so comparing the department and resource grain.
resource_pairs = set(
    zip(
        resource_df['Hospital_id'],
        resource_df['Department_id']
    )
)

#Here we comparing the set difference.
print("Hospital-Department combinations missing in Resource:")
print(department_pairs - resource_pairs)
```

Here we checking the two combination grain is exist in both the datasets to prove is there strong vaildation is available or not.

#### **8.Check Hospital + Department + Date combinations**

```python
#Here we creating a three combination set for department
department_three_pairs = set(
    zip(
        department_df['Hospital_id'],
        department_df['Department_id'],
        department_df['Date']
    )
)

#Here we creating a Three combination Set for resouces 
resource_three_pairs = set(
    zip(
        resource_df['Hospital_id'],
        resource_df['Department_id'],
        resource_df['Date']
    )
)

#Here we checking the set difference to vaildate the combinations in both datasets.
print("Department-Date combinations missing in Resource:")
print(department_three_pairs - resource_three_pairs)
```

Here we are combinely checking whether a combination grain is available in the compraing dataset to strongly vaildate the relationship b/w two sets.
