### **3.CLEAN UP PROCESS BY DUPLICATE REMOVEL AND NULL VALUE FILLING OR REMOVEL.**

Here we will remove the duplicates first and then we will fill the null values presented columns eaither by statistical values, or we can eliminate themn completely.

#### **-->Work on Duplicates**

```python
#Before removing duplicate values we cheching why does the duplication happened whether by different columns values under dept_analytics_id or repeated data.
department_df[department_df['dept_analytics_id'].duplicated(keep=False)].sort_values('dept_analytics_id') # keep =False will give all the dulicate occurences
```

```python
#Removing the dulicate values by the dept_analytics_id.
department_df.drop_duplicates(subset=['dept_analytics_id'], keep='first',inplace=True) #keep='first' will keep the duplicate of first row remove remaining 
                                                                              #subset=['dept_analytics_id'] will used to decide the whether a row is duplicate or not.
```

```python
#Checking whether the column dept_analytics_id have any duplicate values.
department_df['dept_analytics_id'].duplicated().sum()
```

#### **-->WORKING ON NULL VALUES**

```python
#1.Depatment head name will be filled with the frequent item.
department_df['department_head_name']=department_df['department_head_name'].fillna('Dr. Ivan Smith')
```

```python
#2.Here readmission rate pct will first recalculate for the pre precaution.
department_df['readmission_rate_pct']=(department_df['readmission_count'] / department_df['patients_admitted_count'] )*100
department_df['readmission_rate_pct']=department_df['readmission_rate_pct'].fillna(0) #here the when division happened with both zeros a nan will come so we filled with 0.
```

```python
#3.Here we filling the mortality rate pct Na values with the zeros because it was Nulled by the both zeros division.
department_df['mortality_rate_pct']=department_df['mortality_rate_pct'].fillna(0)
```

```python
#4.Here avg length of stay days will be filled with the avg (mean).
department_df['avg_length_of_stay_days']=department_df['avg_length_of_stay_days'].fillna(5.53)
```

```python
#5.Here avg avg_treatment_time_hours will be filled with the avg (mean).
department_df['avg_treatment_time_hours']=department_df['avg_treatment_time_hours'].fillna(67.68)
```

```python
#6.Here avg bed_turnaround_time_hours will be filled with the avg (mean).
department_df['bed_turnaround_time_hours']=department_df['bed_turnaround_time_hours'].fillna(3.20)
```

```python
#7.Here we filling the staff to patient ration by the recalculating again.
department_df['staff_to_patient_ratio']=department_df['nurses_on_duty'] / department_df['occupied_beds_count']
```

```python
#8.Here the avg patient age for every department NA values will  be filled with the avg(mean).
department_df['avg_patient_age']=department_df['avg_patient_age'].fillna(47.8)
```

```python
#9.Here the avg_satisfaction_score for every department NA values will  be filled with the avg(mean).
department_df['avg_satisfaction_score']=department_df['avg_satisfaction_score'].fillna(3.30)
```

```python
#10.Here total billing usd NA will be filled with the avg(mean).
department_df['total_billing_usd']=department_df['total_billing_usd'].fillna(department_df['total_billing_usd'].mean())
```

```python
#11.Here total billing usd NA will be filled with the avg(mean).
department_df['avg_cost_per_patient_usd'] = (
    department_df['avg_cost_per_patient_usd']
    .str.replace('$', '', regex=False) #we replacing the dollar with empty space.
    .str.replace(',', '', regex=False) #we replaing the , with empty space.
    .str.replace('USD','',regex=False) #we replaing the USD with empty space.
    .astype(float)
)
department_df['avg_cost_per_patient_usd']=department_df['avg_cost_per_patient_usd'].fillna(department_df['avg_cost_per_patient_usd'].mean())
```

#### **--> Checking does there exist NAN values**

```python
#Here we only showing the columns  which have null values not all the columns.
Null_values_cols=department_df.isnull().sum() #series datasturcture will be returned here
Null_values_cols[Null_values_cols > 0]
```

### **4.CORRECTION : COL_NAMES, COL_VALUES , DTYPES ETC.**

#### **-->Standardize ID's**

```python
#Here we capitalizing the all the first letter of the column names.
department_df.columns = department_df.columns.str.capitalize()
department_df.columns
```

```python
#Here we Change the numarical ID's into the Strings.
cols=department_df.columns
for ids in cols:
    if ids.endswith('id'):
        department_df[ids]=department_df[ids].astype('string')
```

#### **-->Standardize Departments**

```python
#Here we assigning the values to the department according to the department ids.
Dept_id_mapping={
    'D01':'Cardiology',
    'D02':'Orthopedics',
    'D03':'Neurology',
    'D04':'Oncology',
    'D05':'Pediatrics',
    'D06':'General surgery',
    'D07':'Emergency',
    'D08':'Obstetrics & gynecology',
    'D09':'Nephrology',
    'D10':'Pulmonology',
    'D11':'Gastroenterology',
    'D12':'Psychiatry',
    'D13':'Icu',
    'D14':'General medicine',
    'D15':'Urology'    
}

department_df['Department_name']=department_df['Department_id'].map(Dept_id_mapping)
```

#### **--> Standardize Text**

```python
#Here we are performing removing the leading and trailing spaces from the values in the columns.
column_names=['Hospital_name',
              'Department_type',
              'Record_source_system']

for col in column_names:
    department_df[col]=department_df[col].str.strip()
```

```python
#Here we are performing the capitalize to remove any deplicate values.
column_names_cap=['Hospital_name',
                  'Department_type',
                  'Record_source_system']

for col in column_names_cap:
    department_df[col]=department_df[col].str.capitalize()
```

```python
#Here we are rounding the floating values into 2 decimals.
column_names_float=['Target_occupancy_rate_pct',
                    	'Readmission_rate_pct',
                        'Target_readmission_rate_pct',
                        'Mortality_rate_pct',
                        'Target_avg_los_days',
                        'Avg_treatment_time_hours',
                        'Staff_to_patient_ratio',
                        'Avg_satisfaction_score',
                        'Total_billing_usd',
                        'Avg_cost_per_patient_usd',
                        'Monthly_budget_allocated_usd']

for col in column_names_float:
    department_df[col]=round(department_df[col],2)
```

```python
#Here we are converting the dtype of needed columns into integer.
col_name_int=['Patients_admitted_count',
              'Patients_discharged_count',
              'Readmission_count',
              'Mortality_count',
              'Transfer_events_count',
              'Avg_length_of_stay_days',
              'Avg_patient_age']

for col in col_name_int:
    department_df[col]=department_df[col].astype(int)
```

#### **-->Date standardization**

```python
#here if there are any columns which related date containing different formats we making them as the date related dtype columns
dates_related_cols=['Date','Last_updated_timestamp']
for col in dates_related_cols:
    department_df[col]=pd.to_datetime(department_df[col])
```

#### **-->Outlier validation**

```python
department_df.describe(include='all')
```

```python
#Here we just converted the negative outlier into the positive onces 
department_df['Avg_length_of_stay_days']=abs(department_df['Avg_length_of_stay_days'])
department_df['Avg_treatment_time_hours']=abs(department_df['Avg_treatment_time_hours'])
department_df['Avg_patient_age']=abs(department_df['Avg_patient_age'])
```
