### **3.CLEAN UP PROCESS BY DUPLICATE REMOVEL AND NULL VALUE FILLING OR REMOVEL.**

Here we will remove the duplicates first and then we will fill the null values presented columns eaither by statistical values, or we can eliminate themn completely.

#### -->**Work on Duplicates**

```python
#Before removing duplicate values we cheching why does the duplication happened whether by different columns values under admission id or repeated data.
hospital_df[hospital_df['admission_id'].duplicated(keep=False)].sort_values('admission_id') # keep =False will give all the dulicate occurences
```

```python
#Removing the dulicate values by the admission_id and the returned dataframe will be stored in hospital_df.
hospital_df.drop_duplicates(subset=['admission_id'], keep='first',inplace=True)  #keep='first' will keep the duplicate of first row remove remaining 
                                                                                        #subset=['admission_id'] will used to decide the whether a row is duplicate or not.
```

```python
#Checking whether the new hospital_df column admission_id have any duplicate values.
hospital_df['admission_id'].duplicated().sum()
```

#### **--> WORK ON NULL VALUES**

```python
#1.discharge_date NA values replace with frequent one.
hospital_df['discharge_date']=hospital_df['discharge_date'].fillna(hospital_df['discharge_date'].mode()[0])
```

```python
#2.Room_number Na values replace with frequent one as well and also changed the dtype into int64
hospital_df['room_number']=hospital_df['room_number'].fillna( hospital_df['room_number'].mode()[0])
hospital_df['room_number']=hospital_df['room_number'].astype('int64')
```

```python
#3.Length of stay days  Na values will be replaced with the mean values of that column.
hospital_df['length_of_stay_days']=hospital_df['length_of_stay_days'].fillna(5.0)
hospital_df['length_of_stay_days']=hospital_df['length_of_stay_days'].astype('int64')
```

```python
#4.patient age Na valules will be replaced by the mean value.
hospital_df['patient_age']=hospital_df['patient_age'].fillna(47)
hospital_df['patient_age']=hospital_df['patient_age'].astype('int64')
```

```python
#5.patient gender Na values will be replaced with the others
hospital_df['patient_gender']=hospital_df['patient_gender'].fillna('Others')
hospital_df['patient_gender']=hospital_df['patient_gender'].astype('string')
```

```python
gender_mapping = {
    'M': 'Male',
    'm': 'Male',
    'male': 'Male',
    'Male': 'Male',
    
    'F': 'Female',
    'f': 'Female',
    'female': 'Female',
    'Female': 'Female',
    
    'O': 'Others',
    'Other': 'Others'
}

hospital_df['patient_gender'] = (
    hospital_df['patient_gender']
    .replace(gender_mapping)
)
```

```python
hospital_df['patient_gender'].unique()
```

```python
#6.Attending doctor name replaced with the frequent appeared doctor.
hospital_df['attending_doctor_name']=hospital_df['attending_doctor_name'].fillna('Dr. Daniel Brown')
```

```python
#7.Diagnosis description is marked as Not available for now.
hospital_df['diagnosis_description']=hospital_df['diagnosis_description'].fillna('NOT AVAILABLE')
```

```python
#8.Insurance type Na will be replaced with the frequent value.
hospital_df['insurance_type']=hospital_df['insurance_type'].fillna('Medicare/Medicaid')
```

```python
#Here the Insureance have multiple names under the same type.
insurance_mapping = {
    'MEDICARE/MEDICAID':'Medicare',
    '  Medicare/Medicaid  ':'Medicare',
    'medicare/medicaid':'Medicare',
    'Medicare/Medicaid':'Medicare'

}

hospital_df['insurance_type'] = (
    hospital_df['insurance_type']
    .replace(insurance_mapping)
)
```

```python
#9.Total bill amount values will replaced with the mean and also removel of the doller as well.
#In this column also have doller amount so instead of converting we removed the doller
hospital_df['total_bill_amount'] = pd.to_numeric(
    hospital_df['total_bill_amount']
    .str.replace('$', '', regex=False), # here we are say to the replace method its a normal text not the pattern of a regex
    errors='coerce'  #here we are tell if any invaild data to be encounter make it as null don't push error.
)

hospital_df['total_bill_amount']=hospital_df['total_bill_amount'].fillna(8144.22)
```

```python
#10.IN payment status column we replacing the nan values with frequently appeared value also removed leading and trail spaces by strip.
hospital_df['payment_status'] = (
    hospital_df['payment_status'].str.strip()
)
```

```python
payment_status_mapping = {
'NAN':'Partially Paid'
}

hospital_df['payment_status'] = (
    hospital_df['payment_status']
    .replace(payment_status_mapping)
)

hospital_df['payment_status']=hospital_df['payment_status'].fillna('payment_status')  #Here we filling the nan value with frequently used one.
```

```python
#11.By using mapping we replaced the Na values with the frequently repeated value.
discharge_mapping = {
'nan':'Discharged',
'NAN':'Discharged',
'  nan  ':'Discharged'
}

hospital_df['discharge_status'] = (
    hospital_df['discharge_status']
    .replace(discharge_mapping)
)

hospital_df['discharge_status']=hospital_df['discharge_status'].fillna('Discharged')
```

```python
#12.patient sattisfaction_score Na will be replaced with the mean.
hospital_df['patient_satisfaction_score']=hospital_df['patient_satisfaction_score'].fillna(hospital_df['patient_satisfaction_score'].mean())
```

```python
#13 previous admission id will be filled with the not available .
hospital_df['previous_admission_id']=hospital_df['previous_admission_id'].fillna('NOT AVAILABLE')
```

#### **Diagnosis_description assignment**

```python
#here we findout the diagnosis have not available values to we find out resprective values for diagnosis id's and map to there respective id's by map.
diagnosis_mapping = {
    'A41': 'Sepsis',
    'C50': 'Malignant Neoplasm of Breast',
    'E11': 'Type 2 Diabetes Mellitus',
    'F32': 'Major Depressive Disorder',
    'I10': 'Essential Hypertension',
    'I21': 'Acute Myocardial Infarction',
    'I63': 'Cerebral Infarction (Stroke)',
    'J18': 'Pneumonia',
    'J45': 'Asthma',
    'K35': 'Acute Appendicitis',
    'K80': 'Gallstones',
    'N18': 'Chronic Kidney Disease',
    'N39': 'Urinary Tract Infection',
    'O80': 'Normal Delivery',
    'S72': 'Fracture of Femur'
}

hospital_df['Diagnosis_description']=hospital_df['Diagnosis_code'].map(diagnosis_mapping)
```

#### **Removing Attending doctors duplicate values**

```python
#here we are standardizing the doctor names along side of the doctors id's
doc_uq_ids=hospital_df['Attending_doctor_id'].unique()

doc_uq_names=hospital_df['Attending_doctor_name'].unique()
doc_uq_names=sorted(doc_uq_names)
doc_uq_names=doc_uq_names[:300]

d={'docid':doc_uq_ids,'docname':doc_uq_names}
df=pd.DataFrame(data=d)

hospital_df['Attending_doctor_name'] = hospital_df['Attending_doctor_id'].map(
    df.set_index('docid')['docname']  # set_index will crate a dictionary type values so the map will assign accordingly .
)
```

```python
#Here we are making the negative total bill amount into postive and also fixing the decimal points to the 2.
hospital_df['Total_bill_amount']=round(abs(hospital_df['Total_bill_amount']),2)
```
