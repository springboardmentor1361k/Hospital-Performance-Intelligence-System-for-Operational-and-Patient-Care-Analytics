### **3.CLEAN UP PROCESS BY DUPLICATE REMOVEL AND NULL VALUE FILLING OR REMOVEL.**

Here we will remove the duplicates first and then we will fill the null values presented columns eaither by statistical values, or we can eliminate themn completely.

#### **-->Work on Duplicates**

```python
#Before removing duplicate values we cheching why does the duplication happened whether by different columns values under movement id or repeated data.
patient_df[patient_df['movement_id'].duplicated(keep=False)].sort_values('movement_id') # keep =False will give all the dulicate occurences
```

```python
#Removing the dulicate values by the movement_id.
patient_df.drop_duplicates(subset=['movement_id'], keep='first',inplace=True) #keep='first' will keep the duplicate of first row remove remaining 
                                                                              #subset=['movement_id'] will used to decide the whether a row is duplicate or not.
```

```python
#Checking whether the new hospital_df_dup_rev column movement_id have any duplicate values.
patient_df['movement_id'].duplicated().sum()
```

#### **-->WORKING ON NULL VALUES**

```python
#1.We filled the from department id by using the randow choice function from the numpy.
col = 'from_department_id'

unique_values = patient_df[col].dropna().unique()

mask = patient_df[col].isna()

patient_df.loc[mask, col] = np.random.choice(
    unique_values,
    size=mask.sum()   # here we are tell to the np how may values it has to be generate.here mask.sum()values
)
```

```python
#2.the Na department_name will be replaced by the represented value of the department id.
#why we have to do like this beacuse when we randomly fill the NA values it might one dept id will get verient dept names so thats why,
#we took deptnames mannually and filled according to the dept id.
patient_df['from_department_name']=patient_df['from_department_name'].str.strip()
patient_df['from_department_name']=patient_df['from_department_name'].str.capitalize()
#patient_df['from_department_name'].unique()
patient_df['from_department_id']=patient_df['from_department_id'].astype('string')
```

```python
#Here we displaying all the department names for mannual collection
#pd.set_option('display.max_rows',71)
patient_df[['from_department_id','from_department_name']].drop_duplicates().sort_values(by='from_department_id')
```

```python
Dept_id_mapping={
    'D01':'Cardiology',
    'D02':'Orthopedics',
    'D03':'Neurology',
    'D04':'Oncology',
    'D05':'Pediatrics',
    'D06':'General surgery',
    'D07':'Emergency',
    'D08':'Obstetrics & gynecology',
    'D09':'Nephrology',
    'D10':'Pulmonology',
    'D11':'Gastroenterology',
    'D12':'Psychiatry',
    'D13':'Icu',
    'D14':'General medicine',
    'D15':'Urology'    
}
```

```python
#By the mapping function we get he proper value of the key and assigned to the from department names.
patient_df['from_department_name'] = patient_df['from_department_id'].map(Dept_id_mapping)
patient_df['from_department_name'].unique() # again checking here for any repeated values.
```

```python
#3.Here bed id Na values will be replaced by the frequent id.
patient_df['bed_id']=patient_df['bed_id'].fillna('B0379')
```

```python
#4.Here mode of transport of a patient NA values will be filled by the frequently repeated value.
patient_df['mode_of_transport']=patient_df['mode_of_transport'].fillna('Ambulatory')
```

```python
#5.Here authorized by staff NA value will be replaced by the frequently repeated value.
patient_df['authorized_by_staff']=patient_df['authorized_by_staff'].fillna('Ann Diaz, RN')
```

```python
#6.Here reason for movement column Na value will be filled with the frequent value.
patient_df['reason_for_movement']=patient_df['reason_for_movement'].fillna('Routine discharge')
```

```python
#7.Here duration in department hours will be Na will be filled with the avg.
patient_df['duration_in_department_hours']=patient_df['duration_in_department_hours'].fillna(67.307004)
```

```python
#8.here ward occupancy pct at time column Na values will be replaced with the avg.
patient_df['ward_occupancy_pct_at_time']=patient_df['ward_occupancy_pct_at_time'].fillna(67.428988)
```

#### **--> Checking does there exist NAN values**

```python
#Here we only showing the columns  which have null values not all the columns.
Null_values_cols=patient_df.isnull().sum() #series datasturcture will be returned here
Null_values_cols[Null_values_cols > 0]
```

### **4.CORRECTION : COL_NAMES, COL_VALUES , DTYPES ETC.**

#### **-->Standardize ID's**

```python
#Here we capitalizing the all the first letter of the column names.
patient_df.columns = patient_df.columns.str.capitalize()
patient_df.columns
```

```python
#Here we Change the numarical ID's into the Strings.
cols=patient_df.columns
for ids in cols:
    if ids.endswith('id'):
        patient_df[ids]=patient_df[ids].astype('string')
```

#### **--> Standardize Text**

```python
#Here we removed any leading and trial spaces of the values in the columns.
patient_df['From_location']=patient_df['From_location'].str.strip()
patient_df['Mode_of_transport']=patient_df['Mode_of_transport'].str.strip()
patient_df['Reason_for_movement']=patient_df['Reason_for_movement'].str.strip()
patient_df['Shift']=patient_df['Shift'].str.strip()
```

```python
#Here we capitalize the values in column to reduce the different value repetation. 
patient_df['Hospital_name']=patient_df['Hospital_name'].str.capitalize()
patient_df['Movement_type']=patient_df['Movement_type'].str.capitalize()
patient_df['From_location']=patient_df['From_location'].str.capitalize()
patient_df['To_location']=patient_df['To_location'].str.capitalize()
patient_df['Mode_of_transport']=patient_df['Mode_of_transport'].str.capitalize()
patient_df['Reason_for_movement']=patient_df['Reason_for_movement'].str.capitalize()
patient_df['Shift']=patient_df['Shift'].str.capitalize()
```

#### **-->Standardize Departments**

```python
#Here we assigning the correct department names to the current dept name based on the current dept id.
patient_df['Current_department_name'] = patient_df['Current_department_id'].map(Dept_id_mapping)
```

#### **-->Date standardization**

```python
#here if there are any columns which related date containing different formats we making them as the date related dtype columns
dates_related_cols=['Movement_datetime','Movement_date']
for col in dates_related_cols:
    patient_df[col]=pd.to_datetime(patient_df[col])
```
