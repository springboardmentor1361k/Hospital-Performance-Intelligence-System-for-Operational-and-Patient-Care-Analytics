### **3.CLEAN UP PROCESS BY DUPLICATE REMOVEL AND NULL VALUE FILLING OR REMOVEL.**

Here we will remove the duplicates first and then we will fill the null values presented columns eaither by statistical values, or we can eliminate themn completely.

#### **-->Work on Duplicates**

```python
#Before removing duplicate values we cheching why does the duplication happened whether by different columns values under resource_utilization_id or repeated data.
resource_df[resource_df['resource_utilization_id'].duplicated(keep=False)].sort_values('resource_utilization_id') # keep =False will give all the dulicate occurences
```

```python
#Removing the dulicate values by the 'resource_utilization_id'.
resource_df.drop_duplicates(subset=['resource_utilization_id'], keep='first',inplace=True) #keep='first' will keep the duplicate of first row remove remaining 
                                                                                           #subset=['resource_utilization_id'] will used to decide the whether a row is duplicate or not.
```

```python
#Checking whether the column 'resource_utilization_id' have any duplicate values.
resource_df['resource_utilization_id'].duplicated().sum()
```

#### **-->WORKING ON NULL VALUES**

```python
def mean_cal(val):

    return resource_df[val].sum() / (resource_df[val].count() - resource_df[val].isna().sum())
```

```python
#1.Here we filling the units under maintenance by the original mean (avg) value.
resource_df['units_under_maintenance']=resource_df['units_under_maintenance'].fillna(mean_cal('units_under_maintenance'))
```

```python
#2.Here we recalculating the utilization rate pct to eliminate the any NA values , and rounded into 2 decimals.
resource_df['utilization_rate_pct']=round((resource_df['units_in_use'] / resource_df['total_units_available'])*100,2)
```

```python
#3.Here we filling downtime hours by the original mean (avg) value.
resource_df['downtime_hours']=resource_df['downtime_hours'].fillna(mean_cal('downtime_hours'))
```

```python
#4.Here we filling the nurses on duty snapshot by the original mean (avg) value.
resource_df['nurses_on_duty_snapshot']=resource_df['nurses_on_duty_snapshot'].fillna(mean_cal('nurses_on_duty_snapshot'))
```

```python
#5.Here we filling the doctors on duty snapshot by the original mean (avg) value.
resource_df['doctors_on_duty_snapshot']=resource_df['doctors_on_duty_snapshot'].fillna(mean_cal('doctors_on_duty_snapshot'))
```

```python
#6.Here we filling the overtime hours by the original mean (avg) value.
resource_df['overtime_hours']=resource_df['overtime_hours'].fillna(mean_cal('overtime_hours'))
```

```python
#7.Here we filling the avg response time minutes by the original mean (avg) value.
resource_df['avg_response_time_minutes']=resource_df['avg_response_time_minutes'].fillna(mean_cal('avg_response_time_minutes'))
```

```python
resource_df['cost_per_day_usd'].unique()
```

```python
#8.Here we filling the cost per day usd by the original mean (avg) value.
resource_df['cost_per_day_usd']=(
resource_df['cost_per_day_usd']
.str.replace('$','',regex='False')
.str.replace(',','',regex='False')
.str.lstrip('$')
.str.replace('USD','',regex='False')
.astype(float)
)
resource_df['cost_per_day_usd']=resource_df['cost_per_day_usd'].fillna(mean_cal('cost_per_day_usd'))
```

```python
#9.Here we fill the NA of resource condition based on the frequent item.
resource_df['resource_condition']=resource_df['resource_condition'].fillna('Functional')
```

```python
#10.Here we fill the NA of last maintenance based on the min() item.
resource_df['last_maintenance_date']=resource_df['last_maintenance_date'].astype('date32[pyarrow]')
resource_df['last_maintenance_date']=resource_df['last_maintenance_date'].fillna(resource_df['last_maintenance_date'].min())
```

```python
#11. Here we are filling the planned capacity next month by the mean(avg)
resource_df['planned_capacity_next_month']=resource_df['planned_capacity_next_month'].fillna(mean_cal('planned_capacity_next_month'))
```

### **4.CORRECTION : COL_NAMES, COL_VALUES , DTYPES ETC.**

#### **-->Standardize ID's**

```python
#Here we capitalizing the all the first letter of the column names.
resource_df.columns = resource_df.columns.str.capitalize()
resource_df.columns
```

```python
#Here we Change the numarical ID's into the Strings.
cols=resource_df.columns
for ids in cols:
    if ids.endswith('id'):
        resource_df[ids]=resource_df[ids].astype('string')
```

#### **-->Standardize Departments**

```python
#Here we assigning the values to the department according to the department ids.
Dept_id_mapping={
    'D01':'Cardiology',
    'D02':'Orthopedics',
    'D03':'Neurology',
    'D04':'Oncology',
    'D05':'Pediatrics',
    'D06':'General surgery',
    'D07':'Emergency',
    'D08':'Obstetrics & gynecology',
    'D09':'Nephrology',
    'D10':'Pulmonology',
    'D11':'Gastroenterology',
    'D12':'Psychiatry',
    'D13':'Icu',
    'D14':'General medicine',
    'D15':'Urology'    
}

resource_df['Department_name']=resource_df['Department_id'].map(Dept_id_mapping)
```

#### **-->Date standardization**

```python
#Here we are changing the int dtype of the column date into the date dtype
resource_df['Date']=pd.to_datetime(resource_df['Date'])
resource_df['Last_maintenance_date']=pd.to_datetime(resource_df['Last_maintenance_date']) # here last maintenance date as well.
resource_df['Last_updated_timestamp']=pd.to_datetime(resource_df['Last_updated_timestamp']) # here last updated timestamp date as well.
```

#### **--> Standardize Text**

```python
#Here we are performing removing the leading and trailing spaces from the values in the columns.
column_names=['Resource_type',
              'Resource_category',
              'Resource_condition',
              'Record_source_system'
                                ]

for col in column_names:
    resource_df[col]=resource_df[col].str.strip()
```

```python
#Here we are performing the capitalize to remove any deplicate values.
column_names_cap=['Hospital_name',
                  'Resource_type',
                  'Resource_category',
                  'Resource_condition',
                  'Record_source_system'
                                ]

for col in column_names_cap:
    resource_df[col]=resource_df[col].str.capitalize()
```

```python
#Here we are rounding the floating values into 2 decimals.
column_names_float=['Units_under_maintenance',
                    'Downtime_hours',
                    'Nurses_on_duty_snapshot',
                    'Doctors_on_duty_snapshot',
                    'Overtime_hours',
                    'Avg_response_time_minutes'
                    ]

for col in column_names_float:
    resource_df[col]=round(resource_df[col],2)
```
