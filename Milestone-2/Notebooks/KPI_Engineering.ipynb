```python
hospital_overview = pd.read_csv(
    os.path.join(
        folder,
        "hospital_overview_dataset.csv"
    )
)

# ============================================================
# KPI 4: READMISSION RATE
# ============================================================

# Eligible patients = records where readmission is known
eligible_patients = hospital_overview["readmission"].notna().sum()

# Readmitted patients = readmission value of 1
readmitted_patients = (
    hospital_overview["readmission"] == 1
).sum()

# Calculate rate
readmission_rate = (
    readmitted_patients / eligible_patients
) * 100

print("=" * 60)
print("KPI 4: READMISSION RATE")
print("=" * 60)

print("Eligible Patients:", eligible_patients)
print("Readmitted Patients:", readmitted_patients)
print("Readmission Rate:", round(readmission_rate, 2), "%")
```

```python
department = pd.read_csv(
    os.path.join(folder, "department_analytics_dataset.csv")
)

# Calculate department readmission rate
department["department_readmission_rate"] = (
    department["readmitted_patients"]
    / department["total_admissions"]
) * 100


# Min-Max Normalization
# Higher value = better
department["satisfaction_score"] = (
    (department["average_satisfaction"] -
     department["average_satisfaction"].min())
    /
    (department["average_satisfaction"].max() -
     department["average_satisfaction"].min())
) * 100


# Lower value = better
department["wait_time_score"] = (
    (department["average_wait_time"].max() -
     department["average_wait_time"])
    /
    (department["average_wait_time"].max() -
     department["average_wait_time"].min())
) * 100


# Lower value = better
department["los_score"] = (
    (department["average_length_of_stay"].max() -
     department["average_length_of_stay"])
    /
    (department["average_length_of_stay"].max() -
     department["average_length_of_stay"].min())
) * 100


# Lower value = better
department["readmission_score"] = (
    (department["department_readmission_rate"].max() -
     department["department_readmission_rate"])
    /
    (department["department_readmission_rate"].max() -
     department["department_readmission_rate"].min())
) * 100


# Weighted Department Efficiency Score
department["department_efficiency_score"] = (
    department["satisfaction_score"] * 0.30
    + department["wait_time_score"] * 0.25
    + department["los_score"] * 0.20
    + department["readmission_score"] * 0.25
)


print("=" * 60)
print("KPI 6: DEPARTMENT EFFICIENCY SCORE")
print("=" * 60)

print("\nMinimum Efficiency Score:",
      round(department["department_efficiency_score"].min(), 2))

print("Maximum Efficiency Score:",
      round(department["department_efficiency_score"].max(), 2))

print("Average Efficiency Score:",
      round(department["department_efficiency_score"].mean(), 2))

print("\nTop 5 Departments:")
print(
    department[
        ["department_referral", "department_efficiency_score"]
    ]
    .sort_values("department_efficiency_score", ascending=False)
    .head(5)
    .to_string(index=False)
)
```
