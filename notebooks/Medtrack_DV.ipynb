```python
input_folder = "/content/drive/MyDrive/RAW"
output_folder = "/content/drive/MyDrive/RAW/Cleaned"

os.makedirs(output_folder, exist_ok=True)


# ------------------------------------------------------------
# 2. LOAD DATASETS
# ------------------------------------------------------------

patients = pd.read_csv(
    os.path.join(input_folder, "patients.csv"),
    low_memory=False
)

readmission = pd.read_csv(
    os.path.join(input_folder, "Healthcare Data Analysis for readmission.csv"),
    low_memory=False
)

inpatient = pd.read_csv(
    os.path.join(
        input_folder,
        "Hospital_Inpatient_Discharges__SPARCS_De-Identified___2021_20231012.csv"
    ),
    low_memory=False
)

# DATASET 1: PATIENTS

print("Cleaning patients.csv...")

# Standardize column names
patients.columns = patients.columns.str.strip().str.lower()

# Standardize text columns
patients["name"] = patients["name"].astype(str).str.strip()
patients["service"] = patients["service"].astype(str).str.strip().str.lower()

# Convert dates
patients["arrival_date"] = pd.to_datetime(
    patients["arrival_date"],
    errors="coerce"
)

patients["departure_date"] = pd.to_datetime(
    patients["departure_date"],
    errors="coerce"
)

# Remove exact duplicate rows
patients = patients.drop_duplicates()

# Check invalid values
patients.loc[
    (patients["age"] < 0) | (patients["age"] > 120),
    "age"
] = pd.NA

patients.loc[
    (patients["satisfaction"] < 0) | (patients["satisfaction"] > 100),
    "satisfaction"
] = pd.NA

print("Patients cleaned:", patients.shape)


# DATASET 2: READMISSION / HMIS

print("\nCleaning Healthcare Data Analysis for readmission.csv...")

# Standardize column names
readmission.columns = readmission.columns.str.strip().str.lower()

# Standardize text columns
text_columns = [
    "hospital_name",
    "hospital_ward",
    "patient_gender",
    "patient_race",
    "patient_first_initial",
    "patient_last_name",
    "department_referral",
    "time_slot",
    "doctor_name",
    "doctor_specialty",
    "patient_disease",
    "discharge_status"
]

for col in text_columns:
    readmission[col] = (
        readmission[col]
        .astype(str)
        .str.strip()
    )

# Standardize categorical values
readmission["hospital_ward"] = (
    readmission["hospital_ward"]
    .str.lower()
    .str.strip()
)

readmission["department_referral"] = (
    readmission["department_referral"]
    .str.lower()
    .str.strip()
)

readmission["patient_gender"] = (
    readmission["patient_gender"]
    .str.lower()
    .str.strip()
)

readmission["patient_race"] = (
    readmission["patient_race"]
    .str.lower()
    .str.strip()
)

readmission["discharge_status"] = (
    readmission["discharge_status"]
    .str.lower()
    .str.strip()
)

# Convert dates
date_columns = [
    "admission_date",
    "patient_checkin_date",
    "patient_checkout_date"
]

for col in date_columns:
    readmission[col] = pd.to_datetime(
        readmission[col],
        errors="coerce",
        dayfirst=True
    )

# Remove exact duplicates
readmission = readmission.drop_duplicates()

# Standardize IDs
readmission["hospital_id"] = (
    pd.to_numeric(readmission["hospital_id"], errors="coerce")
    .astype("Int64")
)

readmission["patient_id"] = (
    pd.to_numeric(readmission["patient_id"], errors="coerce")
    .astype("Int64")
)

readmission["doctor_id"] = (
    pd.to_numeric(readmission["doctor_id"], errors="coerce")
    .astype("Int64")
)

# Validate numeric ranges
readmission.loc[
    (readmission["patient_age"] < 0) |
    (readmission["patient_age"] > 120),
    "patient_age"
] = pd.NA

readmission.loc[
    readmission["patient_sat_score"] < 0,
    "patient_sat_score"
] = pd.NA

readmission.loc[
    readmission["patient_length_of_stay"] < 0,
    "patient_length_of_stay"
] = pd.NA

readmission.loc[
    readmission["patient_waittime"] < 0,
    "patient_waittime"
] = pd.NA

# Ensure readmission is 0/1
readmission["readmission"] = pd.to_numeric(
    readmission["readmission"],
    errors="coerce"
)

readmission.loc[
    ~readmission["readmission"].isin([0, 1]),
    "readmission"
] = pd.NA

print("Readmission dataset cleaned:", readmission.shape)


# DATASET 3: INPATIENT DISCHARGES

print("\nCleaning Hospital Inpatient Discharges dataset...")

# Standardize column names
inpatient.columns = (
    inpatient.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

# ------------------------------------------------------------
# Remove exact duplicate rows
# ------------------------------------------------------------

before = len(inpatient)

inpatient = inpatient.drop_duplicates()

after = len(inpatient)

print("Duplicate rows removed:", before - after)


# ------------------------------------------------------------
# Standardize text columns
# ------------------------------------------------------------

text_columns = inpatient.select_dtypes(
    include=["object"]
).columns

for col in text_columns:
    inpatient[col] = (
        inpatient[col]
        .astype("string")
        .str.strip()
    )


# ------------------------------------------------------------
# Convert numeric columns
# ------------------------------------------------------------

numeric_columns = [
    "operating_certificate_number",
    "permanent_facility_id",
    "apr_drg_code",
    "apr_mdc_code",
    "apr_severity_of_illness_code"
]

for col in numeric_columns:
    inpatient[col] = pd.to_numeric(
        inpatient[col],
        errors="coerce"
    )


# ------------------------------------------------------------
# Length of Stay
# ------------------------------------------------------------

inpatient["length_of_stay"] = pd.to_numeric(
    inpatient["length_of_stay"],
    errors="coerce"
)

inpatient.loc[
    inpatient["length_of_stay"] < 0,
    "length_of_stay"
] = pd.NA


# ------------------------------------------------------------
# Financial columns
# Remove commas and convert to numeric
# ------------------------------------------------------------

inpatient["total_charges"] = pd.to_numeric(
    inpatient["total_charges"]
    .astype(str)
    .str.replace(",", "", regex=False),
    errors="coerce"
)

inpatient["total_costs"] = pd.to_numeric(
    inpatient["total_costs"]
    .astype(str)
    .str.replace(",", "", regex=False),
    errors="coerce"
)


# ------------------------------------------------------------
# Birth Weight
# ------------------------------------------------------------

inpatient["birth_weight"] = pd.to_numeric(
    inpatient["birth_weight"],
    errors="coerce"
)


# ------------------------------------------------------------
# Standardize categorical values
# ------------------------------------------------------------

categorical_columns = [
    "gender",
    "race",
    "ethnicity",
    "type_of_admission",
    "patient_disposition",
    "apr_risk_of_mortality",
    "apr_medical_surgical_description",
    "payment_typology_1",
    "payment_typology_2",
    "payment_typology_3",
    "emergency_department_indicator"
]

for col in categorical_columns:
    inpatient[col] = (
        inpatient[col]
        .astype("string")
        .str.strip()
        .str.lower()
    )


# ------------------------------------------------------------
# Missing values
# ------------------------------------------------------------
# Do NOT replace all missing values with 0.
# Keep genuine missing information as NA.
#
# Only numerical financial values are retained as numeric;
# categorical missing values remain missing.
# ------------------------------------------------------------


# ============================================================
# SAVE CLEANED DATASETS
# ============================================================

patients.to_csv(
    os.path.join(output_folder, "patients_cleaned.csv"),
    index=False
)

readmission.to_csv(
    os.path.join(output_folder, "readmission_cleaned.csv"),
    index=False
)

inpatient.to_csv(
    os.path.join(output_folder, "inpatient_discharges_cleaned.csv"),
    index=False
)


# CLEANING SUMMARY

print("\n" + "=" * 70)
print("DATA CLEANING COMPLETED")
print("=" * 70)

print("\nPatients:")
print("Rows:", len(patients))
print("Columns:", len(patients.columns))
print("Duplicates:", patients.duplicated().sum())

print("\nReadmission:")
print("Rows:", len(readmission))
print("Columns:", len(readmission.columns))
print("Duplicates:", readmission.duplicated().sum())

print("\nInpatient Discharges:")
print("Rows:", len(inpatient))
print("Columns:", len(inpatient.columns))
print("Duplicates:", inpatient.duplicated().sum())

print("\nCleaned files saved in:")
print(output_folder)
```

```python
cleaned_folder = "/content/drive/MyDrive/RAW/Cleaned"
standardized_folder = "/content/drive/MyDrive/RAW/Standardized"
os.makedirs(standardized_folder, exist_ok=True)

# 2. LOAD CLEANED DATASETS

patients = pd.read_csv(
    os.path.join(cleaned_folder, "patients_cleaned.csv"),
    low_memory=False
)

readmission = pd.read_csv(
    os.path.join(cleaned_folder, "readmission_cleaned.csv"),
    low_memory=False
)

inpatient = pd.read_csv(
    os.path.join(cleaned_folder, "inpatient_discharges_cleaned.csv"),
    low_memory=False
)

# 3. PATIENT ID STANDARDIZATION

# patients.csv already has IDs like PAT-09484753
patients["patient_id"] = (
    patients["patient_id"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Readmission dataset has numeric patient IDs
readmission["patient_id"] = (
    pd.to_numeric(
        readmission["patient_id"],
        errors="coerce"
    )
    .astype("Int64")
)

# Create a consistent text ID format
readmission["patient_id"] = (
    "PAT-" +
    readmission["patient_id"].astype(str)
)


# ============================================================
# 4. HOSPITAL ID STANDARDIZATION
# ============================================================

readmission["hospital_id"] = (
    pd.to_numeric(
        readmission["hospital_id"],
        errors="coerce"
    )
    .astype("Int64")
)

# Convert hospital ID to consistent text format
readmission["hospital_id"] = (
    "HOSP-" +
    readmission["hospital_id"].astype(str)
)


# ============================================================
# 5. DOCTOR ID STANDARDIZATION
# ============================================================

readmission["doctor_id"] = (
    pd.to_numeric(
        readmission["doctor_id"],
        errors="coerce"
    )
    .astype("Int64")
)

# Convert doctor ID to consistent text format
readmission["doctor_id"] = (
    "DOC-" +
    readmission["doctor_id"].astype(str)
)


# ============================================================
# 6. INPATIENT FACILITY ID STANDARDIZATION
# ============================================================

# Permanent Facility ID
inpatient["permanent_facility_id"] = pd.to_numeric(
    inpatient["permanent_facility_id"],
    errors="coerce"
)

inpatient["permanent_facility_id"] = (
    "FAC-" +
    inpatient["permanent_facility_id"]
    .astype("Int64")
    .astype(str)
)

# Operating Certificate Number
inpatient["operating_certificate_number"] = pd.to_numeric(
    inpatient["operating_certificate_number"],
    errors="coerce"
)

inpatient["operating_certificate_number"] = (
    "CERT-" +
    inpatient["operating_certificate_number"]
    .astype("Int64")
    .astype(str)
)


# ============================================================
# 7. VERIFY ID STANDARDIZATION
# ============================================================

print("=" * 70)
print("ID STANDARDIZATION RESULTS")
print("=" * 70)

print("\nPatients - Patient IDs:")
print(patients["patient_id"].head())

print("\nReadmission - Patient IDs:")
print(readmission["patient_id"].head())

print("\nReadmission - Hospital IDs:")
print(readmission["hospital_id"].head())

print("\nReadmission - Doctor IDs:")
print(readmission["doctor_id"].head())

print("\nInpatient - Facility IDs:")
print(inpatient["permanent_facility_id"].head())

print("\nInpatient - Certificate IDs:")
print(inpatient["operating_certificate_number"].head())


# ============================================================
# 8. SAVE STANDARDIZED DATASETS
# ============================================================

patients.to_csv(
    os.path.join(
        standardized_folder,
        "patients_standardized.csv"
    ),
    index=False
)

readmission.to_csv(
    os.path.join(
        standardized_folder,
        "readmission_standardized.csv"
    ),
    index=False
)

inpatient.to_csv(
    os.path.join(
        standardized_folder,
        "inpatient_discharges_standardized.csv"
    ),
    index=False
)

# 9. FINAL CHECK

print("\n" + "=" * 70)
print("ID STANDARDIZATION COMPLETED")
print("=" * 70)

print("\nPatients:", patients.shape)
print("Readmission:", readmission.shape)
print("Inpatient Discharges:", inpatient.shape)

print("\nStandardized files saved to:")
print(standardized_folder)
```

```python
input_folder = "/content/drive/MyDrive/RAW/Department_Standardized"
output_folder = "/content/drive/MyDrive/RAW/Date_Standardized"

os.makedirs(output_folder, exist_ok=True)

# 1. LOAD DATASETS

patients = pd.read_csv(
    os.path.join(input_folder, "patients_department_standardized.csv")
)

readmission = pd.read_csv(
    os.path.join(input_folder, "readmission_department_standardized.csv")
)

inpatient = pd.read_csv(
    os.path.join(
        input_folder,
        "inpatient_discharges_department_standardized.csv"
    ),
    low_memory=False
)

# 2. PATIENTS DATASET

patients["arrival_date"] = pd.to_datetime(
    patients["arrival_date"],
    errors="coerce"
)

patients["departure_date"] = pd.to_datetime(
    patients["departure_date"],
    errors="coerce"
)

# Create standard date columns
patients["arrival_date"] = patients["arrival_date"].dt.strftime("%Y-%m-%d")
patients["departure_date"] = patients["departure_date"].dt.strftime("%Y-%m-%d")

# 3. READMISSION DATASET

readmission["admission_date"] = pd.to_datetime(
    readmission["admission_date"],
    errors="coerce"
)

readmission["patient_checkin_date"] = pd.to_datetime(
    readmission["patient_checkin_date"],
    errors="coerce"
)

readmission["patient_checkout_date"] = pd.to_datetime(
    readmission["patient_checkout_date"],
    errors="coerce"
)

# Standard date format
readmission["admission_date"] = readmission["admission_date"].dt.strftime("%Y-%m-%d")
readmission["patient_checkin_date"] = readmission["patient_checkin_date"].dt.strftime("%Y-%m-%d")
readmission["patient_checkout_date"] = readmission["patient_checkout_date"].dt.strftime("%Y-%m-%d")


# 4. INPATIENT DISCHARGES DATASET

# Check whether discharge/admission date columns exist
date_columns = [
    col for col in inpatient.columns
    if "date" in col.lower() or "year" in col.lower() # Include 'year' to catch 'discharge_year'
]

print("Inpatient date-related columns:")
print(date_columns)

# Convert any date-related columns to standard format
for col in date_columns:
    if col == "discharge_year":
        # Create a full date string (YYYY-01-01) for proper conversion
        inpatient[col] = pd.to_datetime(
            inpatient[col].astype(str) + '-01-01',
            errors="coerce"
        )
    else:
        inpatient[col] = pd.to_datetime(
            inpatient[col],
            errors="coerce"
        )
    inpatient[col] = inpatient[col].dt.strftime("%Y-%m-%d")

# 5. CHECK DATE VALUES

print("\n" + "="*70)
print("DATE STANDARDIZATION CHECK")
print("="*70)

print("\nPatients:")
print(patients[["arrival_date", "departure_date"]].head())

print("\nReadmission:")
print(
    readmission[
        ["admission_date", "patient_checkin_date", "patient_checkout_date"]
    ].head()
)

print("\nInpatient date columns:")
for col in date_columns:
    print(f"\n{col}:")
    print(inpatient[col].head())

# 6. SAVE DATASETS

patients.to_csv(
    os.path.join(output_folder, "patients_date_standardized.csv"),
    index=False
)

readmission.to_csv(
    os.path.join(output_folder, "readmission_date_standardized.csv"),
    index=False
)

inpatient.to_csv(
    os.path.join(output_folder, "inpatient_discharges_date_standardized.csv"),
    index=False
)


print("\n" + "="*70)
print("DATE STANDARDIZATION COMPLETED")
print("="*70)

print("Patients:", patients.shape)
print("Readmission:", readmission.shape)
print("Inpatient Discharges:", inpatient.shape)

print("\nFiles saved to:")
print(output_folder)
```

```python
input_folder = "/content/drive/MyDrive/RAW/Duplicates_Checked"
output_folder = "/content/drive/MyDrive/RAW/Analytical_Datasets"

os.makedirs(output_folder, exist_ok=True)

# LOAD DATASETS

patients = pd.read_csv(
    os.path.join(
        input_folder,
        "patients_duplicates_checked.csv"
    )
)

readmission = pd.read_csv(
    os.path.join(
        input_folder,
        "readmission_duplicates_checked.csv"
    )
)

inpatient = pd.read_csv(
    os.path.join(
        input_folder,
        "inpatient_discharges_duplicates_checked.csv"
    ),
    low_memory=False
)

# 1. HOSPITAL OVERVIEW DATASET
# Grain: One row = one patient admission
# Source: Readmission dataset

hospital_overview = readmission.copy()

# Select admission-level fields
hospital_overview = hospital_overview[
    [
        "hospital_id",
        "hospital_name",
        "patient_id",
        "admission_date",
        "patient_gender",
        "patient_age",
        "patient_race",
        "patient_disease",
        "patient_length_of_stay",
        "discharge_status",
        "readmission"
    ]
].copy()

# Rename fields for consistency
hospital_overview.rename(
    columns={
        "patient_gender": "gender",
        "patient_age": "age",
        "patient_race": "race",
        "patient_disease": "disease",
        "patient_length_of_stay": "length_of_stay"
    },
    inplace=True
)

hospital_overview.to_csv(
    os.path.join(
        output_folder,
        "hospital_overview_dataset.csv"
    ),
    index=False
)

print("Hospital Overview:", hospital_overview.shape)

# 2. PATIENT FLOW DATASET
# Grain: One row = one patient movement/event
# Source: Readmission dataset

patient_flow = readmission.copy()

patient_flow = patient_flow[
    [
        "hospital_id",
        "hospital_name",
        "patient_id",
        "admission_date",
        "patient_checkin_date",
        "patient_checkout_date",
        "hospital_ward",
        "department_referral",
        "time_slot",
        "patient_waittime",
        "patient_assigned_doctor",
        "discharge_status"
    ]
].copy()

patient_flow.rename(
    columns={
        "patient_waittime": "wait_time",
        "patient_assigned_doctor": "assigned_doctor"
    },
    inplace=True
)

patient_flow.to_csv(
    os.path.join(
        output_folder,
        "patient_flow_dataset.csv"
    ),
    index=False
)

print("Patient Flow:", patient_flow.shape)

# 3. DEPARTMENT ANALYTICS DATASET
# Grain: Hospital + Department + Day
#
# Department referral is the available department-level field.
# Since the readmission dataset contains admission dates,
# aggregate by hospital + department + admission date.

department_analytics = (
    readmission
    .groupby(
        [
            "hospital_id",
            "hospital_name",
            "department_referral",
            "admission_date"
        ],
        dropna=False
    )
    .agg(
        total_admissions=("patient_id", "nunique"),
        average_age=("patient_age", "mean"),
        average_wait_time=("patient_waittime", "mean"),
        average_length_of_stay=("patient_length_of_stay", "mean"),
        average_satisfaction=("patient_sat_score", "mean"),
        readmitted_patients=("readmission", "sum")
    )
    .reset_index()
)

department_analytics.to_csv(
    os.path.join(
        output_folder,
        "department_analytics_dataset.csv"
    ),
    index=False
)

print("Department Analytics:", department_analytics.shape)

# 4. RESOURCE UTILIZATION DATASET
# Grain: Hospital + Department + Day + Resource Type
#
# Available resource fields in the readmission dataset:
# hospital beds available
# occupied beds
#
# We create a daily hospital/ward-level resource table.

resource_utilization = (
    readmission[
        [
            "hospital_id",
            "hospital_name",
            "hospital_ward",
            "admission_date",
            "hospital_beds_available",
            "occupied_beds"
        ]
    ]
    .groupby(
        [
            "hospital_id",
            "hospital_name",
            "hospital_ward",
            "admission_date"
        ],
        dropna=False
    )
    .agg(
        available_beds=("hospital_beds_available", "max"),
        occupied_beds=("occupied_beds", "max")
    )
    .reset_index()
)

# Calculate utilization percentage
resource_utilization["bed_utilization_rate"] = (
    resource_utilization["occupied_beds"]
    / resource_utilization["available_beds"]
    * 100
)

resource_utilization.to_csv(
    os.path.join(
        output_folder,
        "resource_utilization_dataset.csv"
    ),
    index=False
)

print("Resource Utilization:", resource_utilization.shape)

# FINAL VALIDATION

print("\n" + "="*70)
print("FOUR ANALYTICAL DATASETS CREATED")
print("="*70)

print("\n1. Hospital Overview")
print(hospital_overview.head())

print("\n2. Patient Flow")
print(patient_flow.head())

print("\n3. Department Analytics")
print(department_analytics.head())

print("\n4. Resource Utilization")
print(resource_utilization.head())

print("\n" + "="*70)
print("FILES SAVED TO")
print("="*70)

print(output_folder)
```

```python
folder = "/content/drive/MyDrive/RAW/Analytical_Datasets"

# Load datasets
overview = pd.read_csv(
    os.path.join(folder, "hospital_overview_dataset.csv")
)

resource = pd.read_csv(
    os.path.join(folder, "resource_utilization_dataset.csv")
)

department = pd.read_csv(
    os.path.join(folder, "department_analytics_dataset.csv")
)

# ---------------- KPI 1 ----------------
total_admissions = overview["patient_id"].count()

# ---------------- KPI 2 ----------------
total_beds = overview["hospital_beds_available"] if "hospital_beds_available" in overview.columns else None

# Use resource dataset for bed calculation
available_beds = resource["available_beds"].sum()
occupied_beds = resource["occupied_beds"].sum()

occupancy_rate = (occupied_beds / available_beds) * 100

# ---------------- KPI 3 ----------------
average_los = overview["length_of_stay"].mean()

# ---------------- KPI 4 ----------------
eligible_patients = overview["readmission"].notna().sum()
readmitted_patients = (overview["readmission"] == 1).sum()

readmission_rate = (
    readmitted_patients / eligible_patients
) * 100

# ---------------- KPI 5 ----------------
bed_utilization_rate = (
    occupied_beds / available_beds
) * 100

# ---------------- KPI 6 ----------------
department["department_readmission_rate"] = (
    department["readmitted_patients"]
    / department["total_admissions"]
) * 100

department["satisfaction_score"] = (
    (department["average_satisfaction"]
     - department["average_satisfaction"].min())
    /
    (department["average_satisfaction"].max()
     - department["average_satisfaction"].min())
) * 100

department["wait_time_score"] = (
    (department["average_wait_time"].max()
     - department["average_wait_time"])
    /
    (department["average_wait_time"].max()
     - department["average_wait_time"].min())
) * 100

department["los_score"] = (
    (department["average_length_of_stay"].max()
     - department["average_length_of_stay"])
    /
    (department["average_length_of_stay"].max()
     - department["average_length_of_stay"].min())
) * 100

department["readmission_score"] = (
    (department["department_readmission_rate"].max()
     - department["department_readmission_rate"])
    /
    (department["department_readmission_rate"].max()
     - department["department_readmission_rate"].min())
) * 100

department["department_efficiency_score"] = (
    department["satisfaction_score"] * 0.30
    + department["wait_time_score"] * 0.25
    + department["los_score"] * 0.20
    + department["readmission_score"] * 0.25
)

efficiency_average = department["department_efficiency_score"].mean()


# ---------------- VALIDATION TABLE ----------------

validation = pd.DataFrame({
    "KPI": [
        "Total Admissions",
        "Occupancy Rate",
        "Average Length of Stay",
        "Readmission Rate",
        "Bed Utilization Rate",
        "Department Efficiency Score"
    ],

    "Calculated Value": [
        total_admissions,
        round(occupancy_rate, 2),
        round(average_los, 2),
        round(readmission_rate, 2),
        round(bed_utilization_rate, 2),
        round(efficiency_average, 2)
    ],

    "Formula / Definition": [
        "Count of admission records",
        "Occupied Beds / Available Beds × 100",
        "Average Length of Stay",
        "Readmitted Patients / Eligible Patients × 100",
        "Occupied Beds / Available Beds × 100",
        "Weighted normalized score: Satisfaction 30%, Wait Time 25%, LOS 20%, Readmission 25%"
    ]
})

print("=" * 80)
print("KPI VALIDATION TABLE")
print("=" * 80)

print(validation.to_string(index=False))
```
